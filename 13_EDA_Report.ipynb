## Statistical Findings

Statistical findings summarize the main characteristics of numerical data.

They help us understand typical values, variability, distribution patterns, and unusual observations.

The main areas covered are:

- Central Tendency: Mean, median, and mode.
- Dispersion: Range, variance, and standard deviation.
- Distribution: The shape and spread of numerical values.
- Outliers: Values that differ substantially from the general data pattern.

### Objective
Summarize the statistical characteristics of the retail dataset and document findings that may be relevant to further analysis.



# Step 2: Demonstrate the Concept

## Simple Business Example

Consider the following example of product prices.

| Price | Value |
|---|---:|
| Product A | 100 |
| Product B | 120 |
| Product C | 130 |
| Product D | 150 |
| Product E | 500 |

These values are illustrative examples, not actual results from the retail dataset.

### Statistical Understanding

- Mean represents the arithmetic average.
- Median represents the middle value after sorting.
- Range measures the difference between the maximum and minimum.
- Standard deviation describes how widely values vary around the mean.
- Distribution analysis helps reveal skewness and concentration.
- Outlier analysis identifies unusually distant values for further investigation.

An unusual value may be valid, so statistical findings should be interpreted in context.


```python

# Select numerical features for statistical analysis
numeric_df = df.select_dtypes(include="number")

# Calculate central tendency and dispersion
statistical_summary = pd.DataFrame({
    "Mean": numeric_df.mean(),
    "Median": numeric_df.median(),
    "Mode": numeric_df.mode().iloc[0],
    "Minimum": numeric_df.min(),
    "Maximum": numeric_df.max(),
    "Range": numeric_df.max() - numeric_df.min(),
    "Variance": numeric_df.var(),
    "Standard_Deviation": numeric_df.std(),
    "Skewness": numeric_df.skew()
})

print("Statistical Summary:")
display(statistical_summary.round(3))

# Identify potential outliers using the IQR rule
outlier_summary = []

for column in numeric_df.columns:
    values = numeric_df[column].dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outlier_count = (
        (values < lower_limit) | (values > upper_limit)
    ).sum()

    outlier_summary.append({
        "Feature": column,
        "Lower_Limit": lower_limit,
        "Upper_Limit": upper_limit,
        "Potential_Outlier_Count": outlier_count
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

print("\nPotential Outliers Using the IQR Rule:")
display(outlier_summary_df.round(3))
```


# Step 4: Interpretation and Report

## Code Explanation

- `select_dtypes(include="number")` selects numerical columns.
- `mean()`, `median()`, and `mode()` calculate central tendency.
- `min()`, `max()`, and `max() - min()` describe the range.
- `var()` and `std()` calculate variance and standard deviation.
- `skew()` measures distribution asymmetry.
- The IQR rule identifies potential outliers using quartiles and calculated limits.

## Statistical Findings

Use the actual output to document:

# Step 2: Demonstrate the Concept

## Simple Business Example

Consider the following illustrative retail data.

| Quantity | Price | Sales Value |
|---:|---:|---:|
| 1 | 100 | 100 |
| 2 | 100 | 200 |
| 3 | 100 | 300 |

These are example values, not actual results from the dataset.

### Understanding the Relationship

- Sales Value changes when Quantity changes, while Price remains constant.
- Correlation can help summarize the linear relationship between numerical variables.
- Relationships should be interpreted in the context of how each feature is calculated.

Correlation alone does not prove that one variable causes a change in another.


```python

import numpy as np

# Select numerical features
numeric_df = df.select_dtypes(include="number")

# Calculate the correlation matrix
correlation_matrix = numeric_df.corr()

print("Correlation Matrix:")
display(correlation_matrix.round(3))

# Identify strongly correlated feature pairs
correlation_threshold = 0.7

correlation_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
).stack().reset_index()

correlation_pairs.columns = [
    "Feature_1", "Feature_2", "Correlation"
]

strong_relationships = correlation_pairs[
    correlation_pairs["Correlation"].abs() >= correlation_threshold
].sort_values(
    "Correlation",
    key=lambda values: values.abs(),
    ascending=False
)

print("\nFeature Pairs with Absolute Correlation >= 0.7:")
display(strong_relationships)

print("\nNumerical Feature Summary:")
display(numeric_df.describe().round(2))

```


# Step 4: Interpretation and Report

## Code Explanation

- `select_dtypes(include="number")` selects numerical features.
- `corr()` calculates Pearson correlation coefficients.
- The upper-triangle filter avoids displaying each feature pair twice.
- The absolute correlation threshold identifies pairs with a correlation magnitude of at least 0.7.
- `describe()` summarizes numerical feature distributions.

## Relationship Findings

Use the actual output to document:
- Feature pairs with stronger positive correlations.
- Feature pairs with stronger negative correlations.
- Numerical features with weaker linear relationships.
- Important patterns that agree with previous EDA findings.

## Business Interpretation

Relationships between features can help explain patterns in retail transactions and identify features that may be useful for later analysis.

Correlation does not establish causation. Results can also be influenced by returns, unusual values, and the characteristics of the dataset.

## Step 2 – Demonstrate the Concept

Consider an online retail dataset containing transaction quantities, prices, invoice dates, and countries.

Examples of observations from visualizations include:

- A histogram can show whether numerical values are concentrated in a particular range.
- A box plot can help identify potential outliers.
- A bar chart can compare recorded sales values across countries.

These observations should be based on the actual charts and their outputs. A visualization may reveal a possible pattern, but further investigation may be required before drawing a business conclusion.


```python

# Summarize key numerical and business patterns for the EDA report

# Calculate transaction-row sales value
df["SalesValue"] = df["Quantity"] * df["Price"]

# Prepare a summary of numerical feature distributions
numeric_summary = df[
    ["Quantity", "Price", "SalesValue"]
].describe().T

numeric_summary["Skewness"] = df[
    ["Quantity", "Price", "SalesValue"]
].skew()

print("Numerical Distribution Summary:")
display(numeric_summary.round(2))

# Identify potential outliers using the IQR method
outlier_summary = []

for column in ["Quantity", "Price", "SalesValue"]:
    values = df[column].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (values < lower_bound) | (values > upper_bound)
    ).sum()

    outlier_summary.append({
        "Feature": column,
        "Potential_Outlier_Count": int(outlier_count)
    })

print("\nPotential Outliers Identified in Numerical Features:")
display(pd.DataFrame(outlier_summary))

# Summarize recorded sales values by country
country_summary = (
    df.groupby("Country", dropna=False)
    .agg(
        Transaction_Rows=("Invoice", "count"),
        Total_Quantity=("Quantity", "sum"),
        Recorded_Sales_Value=("SalesValue", "sum")
    )
    .sort_values("Recorded_Sales_Value", ascending=False)
)

print("\nTop 10 Countries by Recorded Sales Value:")
display(country_summary.head(10).round(2))

# Summarize recorded sales values by month
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"], errors="coerce"
)

monthly_summary = (
    df.dropna(subset=["InvoiceDate"])
    .assign(
        Month=lambda data: data["InvoiceDate"].dt.to_period("M")
    )
    .groupby("Month")
    .agg(
        Transaction_Rows=("Invoice", "count"),
        Recorded_Sales_Value=("SalesValue", "sum")
    )
    .reset_index()
)

monthly_summary["Month"] = monthly_summary["Month"].astype(str)

print("\nMonthly Recorded Sales Summary:")
display(monthly_summary)
```


## Step 4 – Interpretation / Report

Review the outputs and document the actual findings from the visualizations and summary tables.

- **Distribution:** Use the numerical summary and skewness values to describe the distributions of Quantity, Price, and SalesValue.
- **Outliers:** Use the IQR summary to identify features containing potential outliers. These values require investigation and should not be removed automatically.
- **Country comparison:** Use the country summary to identify countries with higher or lower recorded sales values.
- **Time trends:** Use the monthly summary to identify increases, decreases, or fluctuations in recorded sales values.
- **Business interpretation:** Treat these results as observations that require further investigation before making business decisions.

Note: SalesValue is calculated as Quantity multiplied by Price for each transaction row. Negative quantities may affect the totals, so these values should not automatically be interpreted as final net revenue.

| Potential problem | Possible impact |
|---|---|
| Missing Customer ID | Customer-level analysis may be limited. |
| Duplicate records | Transaction summaries may be affected. |

```python

# Identify potential data problems for the EDA report

# 1. Missing values
missing_summary = df.isnull().sum().reset_index()
missing_summary.columns = ["Feature", "Missing_Count"]

missing_summary["Missing_Percentage"] = (
    missing_summary["Missing_Count"] / len(df) * 100
)

print("Missing Values:")
display(
    missing_summary[missing_summary["Missing_Count"] > 0]
    .sort_values("Missing_Count", ascending=False)
    .round(2)
)

# 2. Duplicate records
duplicate_count = df.duplicated().sum()

print("Exact Duplicate Rows:", duplicate_count)

# 3. Potentially unusual numerical values
value_checks = pd.DataFrame({
    "Check": [
        "Negative Quantity",
        "Zero Quantity",
        "Negative Price",
        "Zero Price"
    ],
    "Count": [
        (df["Quantity"] < 0).sum(),
        (df["Quantity"] == 0).sum(),
        (df["Price"] < 0).sum(),
        (df["Price"] == 0).sum()
    ]
})

print("\nUnusual Numerical Values:")
display(value_checks)

# 4. Data type inspection
dtype_summary = pd.DataFrame({
    "Feature": df.columns,
    "Data_Type": df.dtypes.astype(str).values
})

print("\nData Types:")
display(dtype_summary)

# 5. Potential date issues
invoice_dates = pd.to_datetime(
    df["InvoiceDate"], errors="coerce"
)

invalid_date_count = invoice_dates.isna().sum()

print("\nMissing or Invalid Invoice Dates:", invalid_date_count)

# 6. Potential outliers using the IQR method
outlier_results = []

for column in ["Quantity", "Price"]:
    values = df[column].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    count = (
        (values < lower_bound) | (values > upper_bound)
    ).sum()

    outlier_results.append({
        "Feature": column,
        "Potential_Outliers": int(count)
    })

print("\nPotential Outliers:")
display(pd.DataFrame(outlier_results))
```
