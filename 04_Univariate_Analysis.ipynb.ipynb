# Numerical Variables

## Topic 1: Mean

### Step 1: Understand the Concept

Mean is the arithmetic average of all numerical values in a variable. It is calculated by adding all valid values and dividing the sum by the number of valid values.

**Formula:**

Mean = Sum of all values / Number of valid values

**Why do we use Mean?**

- To understand the average value of a numerical variable.
- To summarize a large dataset using one number.
- To understand the general level of the data.
- To support exploratory data analysis and feature understanding.

**Important:** Mean can be affected by extremely high or low values. Therefore, it should be interpreted along with the distribution of the same variable.

## Step 2: Demonstrate the Concept

**Simple Example**

Consider these quantity values:

| Quantity |
|---:|
| 2 |
| 4 |
| 6 |
| 8 |

Mean = (2 + 4 + 6 + 8) / 4 = 5

**Real-World Example**

If a shop sells 5, 10, and 15 items across three transactions, the average quantity per transaction is 10 items.

**Business Example**

A retailer can use the mean quantity to understand the average number of items recorded per transaction.

**AI/ML Use Case**

Mean helps describe numerical features during exploratory data analysis and can be used for mean imputation when appropriate.

**Importance in Data Analysis**

Mean provides a quick summary of a numerical variable. However, unusual values may increase or decrease the average, so the result must be interpreted carefully.

**Univariate Analysis Rule:** In this topic, we analyze only the `Quantity` variable. We do not compare it with other columns.

```python
import pandas as pd
import numpy as np

file_path = "online_retail_II.xlsx"
df = pd.read_excel(
    file_path,
    sheet_name="Year 2009-2010"
)

print("Dataset loaded successfully!")
display(df.head())
```

```python
# Select the Quantity variable only
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

# Calculate the mean, excluding missing values
quantity_mean = quantity.mean()

print("Variable: Quantity")
print("Mean Quantity:", quantity_mean)
```

## Code Explanation

**1. `df["Quantity"]`**

Selects only the `Quantity` column from the Online Retail II dataset. This follows the univariate analysis rule.

**2. `pd.to_numeric(df["Quantity"], errors="coerce")`**

Converts the column into numeric values. Values that cannot be converted become missing values (`NaN`).

**Why is it used?**

Mean calculation requires numerical data. This conversion helps prevent errors if unexpected text values exist.

**3. `quantity.mean()`**

Calculates the arithmetic mean of the Quantity values. By default, pandas skips missing values.

**4. `print()`**

Displays the variable name and its calculated mean.

**How to interpret the output**

- A positive mean indicates that the average recorded quantity is positive.
- A mean near zero may indicate that positive and negative quantities offset each other.
- A negative mean indicates that the average recorded quantity is negative.

In the Online Retail II dataset, negative quantities can occur in cancellation or return records. Therefore, the mean represents the average of the recorded quantities, not necessarily the average number of items sold in completed sales.

**Finding**

Record the numerical output displayed by your notebook. Do not assume a result before running the code.

**Impact on the ML Pipeline**

Mean helps summarize a numerical feature during exploratory data analysis. It may also be used for missing-value imputation when suitable, but the imputation strategy should be chosen based on the data distribution and business meaning.

# Topic 2: Median

## Step 1: Understand the Concept

The median is the middle value of a numerical dataset after arranging the values in ascending or descending order.

- If the number of observations is odd, the median is the middle value.
- If the number of observations is even, the median is the average of the two middle values.

**Formula:**

For an even number of observations:

Median = (Lower Middle Value + Upper Middle Value) / 2

**Why do we use Median?**

- To identify the central value of a numerical variable.
- To understand a typical value in the dataset.
- To summarize data that may contain extreme values.
- To support exploratory data analysis.

The median is generally less affected by extreme values than the mean.

## Step 2: Demonstrate the Concept

**Simple Example**

Consider the following quantity values:

| Quantity |
|---:|
| 2 |
| 4 |
| 6 |
| 8 |
| 10 |

The middle value is 6.

**Median = 6**

For an even number of observations, such as 2, 4, 6, and 8, the median is (4 + 6) / 2 = 5.

**Real-World Example**

The median waiting time helps identify the middle waiting time when customers' waiting times are arranged in order.

**Business Example**

A retailer can use the median transaction quantity to understand the central quantity without allowing a few unusually large orders to dominate the summary.

**AI/ML Use Case**

The median is often used to fill missing numerical values when the feature contains extreme values or has a skewed distribution.

**Importance in Data Analysis**

The median helps summarize the center of a numerical variable and is useful when the data contains unusually high or low observations.

**Univariate Analysis Rule:** Analyze only the `Quantity` variable in this topic. Do not compare it with other variables.

```python
# Select the Quantity variable only
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

# Calculate the median, excluding missing values
quantity_median = quantity.median()

print("Variable: Quantity")
print("Median Quantity:", quantity_median)
```

## Code Explanation

**1. `df["Quantity"]`**

Selects the `Quantity` column from the dataset. No other variable is analyzed.

**2. `pd.to_numeric(df["Quantity"], errors="coerce")`**

Converts the column into numeric values. Any value that cannot be converted becomes `NaN`.

**3. `quantity.median()`**

Calculates the middle value of the available numerical observations. Pandas automatically excludes missing values from this calculation.

**4. `print()`**

Displays the variable name and its calculated median.

**How to Interpret the Output**

- The median represents the central recorded quantity.
- If the median is positive, the middle observation is positive.
- If the median is zero, the middle observation is zero.
- If the median is negative, the middle observation is negative.

The Online Retail II dataset may contain negative quantities for cancellations or returns. Therefore, interpret the median as the center of the recorded quantities, not automatically as the typical quantity of a completed sale.

# Topic 3: Mode

## Step 1: Understand the Concept

The mode is the value that occurs most frequently in a dataset.

A dataset can have:
- **One mode:** One value occurs more frequently than the others.
- **Multiple modes:** Two or more values share the highest frequency.
- **No unique mode:** All values occur with the same frequency.

**Why do we use Mode?**

- To identify the most frequently occurring value.
- To understand common patterns in numerical data.
- To summarize frequently repeated observations.
- To support exploratory data analysis.

Mode can be used with both numerical and categorical variables. In this topic, we analyze only the numerical `Quantity` variable.

## Step 2: Demonstrate the Concept

**Simple Example**

| Quantity |
|---:|
| 2 |
| 4 |
| 4 |
| 4 |
| 6 |
| 8 |

The value `4` occurs three times, more often than any other value.

**Mode = 4**

**Real-World Example**

If customers repeatedly order the same number of items, the most frequent quantity can help identify a common order size.

**Business Example**

A retailer can examine the most frequently recorded transaction quantity to understand common purchasing patterns.

**AI/ML Use Case**

Mode can be used to fill missing values in numerical or categorical features when the most frequent value is a suitable choice.

**Importance in Data Analysis**

Mode helps identify the most common observation. Unlike the mean, it does not depend on adding all values together.

**Univariate Analysis Rule:** Analyze only the `Quantity` variable. Do not compare it with another column.

```python
# Select the Quantity variable only
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

# Calculate the mode, excluding missing values
quantity_mode = quantity.mode()

print("Variable: Quantity")
print("Mode Quantity:")

if quantity_mode.empty:
    print("No mode available.")
else:
    display(quantity_mode.to_frame(name="Mode"))
```

## Code Explanation

**1. `df["Quantity"]`**

Selects only the `Quantity` column from the Online Retail II dataset.

**2. `pd.to_numeric(df["Quantity"], errors="coerce")`**

Converts the column to numeric values. Values that cannot be converted become `NaN`.

**3. `quantity.mode()`**

Finds the value or values that occur most frequently. Pandas returns a Series because multiple values can share the highest frequency. Missing values are excluded by default.

**4. `quantity_mode.empty`**

Checks whether the result contains any mode values.

**5. `display(quantity_mode.to_frame(name="Mode"))`**

Displays the mode result as a table.

**How to Interpret the Output**

- One row means one mode was identified.
- Multiple rows mean multiple values share the highest frequency.
- An empty result means no mode value was returned, for example, if no valid observations are available.

**Finding**

Run the code and note the actual mode value or values displayed. Do not assume the most frequent quantity before checking the output.

**Impact on the ML Pipeline**

Mode can help describe frequently occurring feature values and may be useful for imputation. However, the chosen imputation method should match the feature's distribution and business meaning.

# Topic 6: Range

## Step 1: Understand the Concept

Range is the difference between the maximum and minimum values of a numerical variable.

**Formula:**

Range = Maximum - Minimum

**Why do we use Range?**

- To understand the overall spread of numerical values.
- To identify the distance between the smallest and largest observations.
- To summarize the variability of a variable.
- To support exploratory data analysis.

Range depends only on the minimum and maximum values. Therefore, an extreme observation can strongly affect it.

## Step 2: Demonstrate the Concept

**Simple Example**

| Quantity |
|---:|
| 2 |
| 4 |
| 6 |
| 8 |
| 10 |

Maximum = 10

Minimum = 2

Range = 10 - 2 = **8**

**Real-World Example**

The range of daily temperatures is the difference between the highest and lowest temperatures recorded that day.

**Business Example**

A retailer can use the range to understand the spread between the smallest and largest recorded order quantities.

**AI/ML Use Case**

Range helps describe feature variability and provides context when selecting scaling methods for numerical features.

**Importance in Data Analysis**

A large range indicates a wide distance between the extreme values. However, it does not explain how values are distributed between those extremes.

**Univariate Analysis Rule:** Calculate the range of `Quantity` only. Do not compare it with another variable.

```python
# Select the Quantity variable only
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

# Calculate the minimum and maximum
quantity_minimum = quantity.min()
quantity_maximum = quantity.max()

# Calculate the range
quantity_range = quantity_maximum - quantity_minimum

print("Variable: Quantity")
print("Minimum Quantity:", quantity_minimum)
print("Maximum Quantity:", quantity_maximum)
print("Range of Quantity:", quantity_range)
```

## Code Explanation

**1. `df["Quantity"]`**

Selects only the `Quantity` column for analysis.

**2. `pd.to_numeric(df["Quantity"], errors="coerce")`**

Converts the column into numeric values. Values that cannot be converted become `NaN`.

**3. `quantity.min()`**

Returns the smallest valid quantity value.

**4. `quantity.max()`**

Returns the largest valid quantity value.

**5. `quantity_maximum - quantity_minimum`**

Calculates the range by subtracting the minimum from the maximum.

**6. `print()`**

Displays the minimum, maximum, and calculated range.

**How to Interpret the Output**

- A range of zero means all valid observations have the same value.
- A larger range means the distance between the smallest and largest observations is greater.
- A very large range may indicate extreme observations that require further investigation.

**Finding**

Run the code and record the actual range. In the Online Retail II dataset, negative quantities may occur in return or cancellation records, so the range may span both negative and positive values.

**Impact on the ML Pipeline**

# Topic 7: Variance

## Step 1: Understand the Concept

Variance measures how far numerical values are spread around their mean.

It is calculated by finding the deviations from the mean, squaring those deviations, and averaging them.

**Sample Variance Formula:**

Sample Variance = Sum of Squared Deviations / (Number of Observations - 1)

**Why do we use Variance?**

- To measure the spread of numerical values.
- To understand how much observations vary around the mean.
- To identify features with high variability.
- To support statistical analysis and exploratory data analysis.

Variance is expressed in squared units. Therefore, it can be less intuitive to interpret than standard deviation.

## Step 2: Demonstrate the Concept

**Simple Example**

Consider the quantity values 2, 4, and 6.

| Quantity | Deviation from Mean (4) | Squared Deviation |
|---:|---:|---:|
| 2 | -2 | 4 |
| 4 | 0 | 0 |
| 6 | 2 | 4 |

The population variance is (4 + 0 + 4) / 3 = 2.67.

The sample variance is (4 + 0 + 4) / (3 - 1) = **4**.

**Real-World Example**

Variance can describe how much daily delivery times vary around their average.

**Business Example**

A retailer can use variance to understand how widely recorded order quantities are spread around their average.

**AI/ML Use Case**

Variance helps identify numerical features with different levels of variability. Very low-variance features may provide limited information for some models.

**Importance in Data Analysis**

Variance quantifies the spread of values rather than just showing their minimum and maximum.

**Univariate Analysis Rule:** Analyze only the `Quantity` variable. Do not compare it with another column.

```python
# Select the Quantity variable only
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

# Calculate the sample variance, excluding missing values
quantity_variance = quantity.var()

print("Variable: Quantity")
print("Sample Variance of Quantity:", quantity_variance)
```

## Code Explanation

**1. `df["Quantity"]`**

Selects only the `Quantity` column from the dataset.

**2. `pd.to_numeric(df["Quantity"], errors="coerce")`**

Converts the selected column into numeric values. Values that cannot be converted become `NaN`.

**3. `quantity.var()`**

Calculates the sample variance of the available quantity values.

By default, pandas uses `ddof=1`, so it divides the sum of squared deviations by the number of valid observations minus one. Missing values are excluded.

**4. `print()`**

Displays the variable name and calculated sample variance.

**How to Interpret the Output**

- A variance of zero means all valid observations have the same value.
- A larger variance indicates greater spread around the mean, though the magnitude depends on the units and scale.
- Variance cannot be negative.

**Finding**

Run the code and record the actual result. Because this dataset may contain large positive and negative quantity values, the variance can be strongly influenced by extreme observations.

**Impact on the ML Pipeline**

Variance helps describe feature variability and can support feature screening. However, a high variance does not automatically make a feature useful, and a low variance does not always mean it should be removed.

# Topic 8: Standard Deviation

## Step 1: Understand the Concept

Standard deviation measures how much numerical values typically vary around their mean.

It is the square root of the variance and is expressed in the same units as the original variable.

**Formula:**

Standard Deviation = Square Root of Variance

**Why do we use Standard Deviation?**

- To understand the spread of numerical values.
- To identify whether observations are close to or far from the mean.
- To describe the variability of a feature.
- To support exploratory data analysis and statistical analysis.

A small standard deviation indicates less spread, while a large standard deviation indicates greater spread. Interpretation depends on the data and its scale.

## Step 2: Demonstrate the Concept

**Simple Example**

Consider the quantity values 2, 4, and 6.

| Quantity | Deviation from Mean (4) |
|---:|---:|
| 2 | -2 |
| 4 | 0 |
| 6 | 2 |

The population variance is approximately 2.67.

Population standard deviation = √2.67 ≈ **1.63**

**Real-World Example**

Standard deviation can describe how much delivery times vary around their average.

**Business Example**

A retailer can use standard deviation to understand the variability of recorded order quantities.

**AI/ML Use Case**

Standard deviation helps understand feature distributions and is useful when standardizing numerical features for algorithms that are sensitive to feature scales.

**Importance in Data Analysis**

Standard deviation describes spread in the original unit, making it easier to interpret than variance.

**Univariate Analysis Rule:** Analyze only the `Quantity` variable. Do not compare it with other columns.

```python
# Select the Quantity variable only
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

# Calculate the sample standard deviation
quantity_std = quantity.std()

print("Variable: Quantity")
print("Sample Standard Deviation of Quantity:", quantity_std)
```

## Code Explanation

**1. `df["Quantity"]`**

Selects only the `Quantity` column from the dataset.

**2. `pd.to_numeric(df["Quantity"], errors="coerce")`**

Converts the column into numeric values. Values that cannot be converted become `NaN`.

**3. `quantity.std()`**

Calculates the sample standard deviation. By default, pandas uses `ddof=1`, dividing by the number of valid observations minus one when calculating variance before taking its square root. Missing values are excluded.

**4. `print()`**

Displays the variable name and calculated standard deviation.

**How to Interpret the Output**

- A standard deviation of zero means all valid observations have the same value.
- A larger value indicates greater spread around the mean.
- Standard deviation is non-negative and uses the same units as `Quantity`.

**Finding**

Run the code and record the actual result. The Online Retail II dataset may contain extreme positive or negative quantities, which can affect the standard deviation.

**Impact on the ML Pipeline**

Standard deviation helps describe numerical features and supports scaling methods such as standardization. Always consider the feature's distribution and business meaning before choosing a preprocessing method.

# Topic 9: Quartiles

## Step 1: Understand the Concept

Quartiles are statistical measures that divide an ordered numerical dataset into four parts. They help us understand how the values are distributed within a single variable.

There are three main quartiles:

- **Q1 (First Quartile):** The 25th percentile. Approximately 25% of the values fall at or below this point.
- **Q2 (Second Quartile):** The 50th percentile, also called the median. Approximately 50% of the values fall at or below this point.
- **Q3 (Third Quartile):** The 75th percentile. Approximately 75% of the values fall at or below this point.

Quartiles are useful for understanding the distribution and spread of a numerical variable. The exact quartile values depend on the dataset and the calculation method.

## Step 2: Demonstrate the Concept

### Simple Example

Consider the following ordered values of a single variable:

| Quantity Values | 10 | 20 | 30 | 40 | 50 | 60 | 70 |
|---|---:|---:|---:|---:|---:|---:|---:|

Quartiles help us identify the lower, middle, and upper portions of these values. The actual quartile values are calculated from the dataset.

### Real-World Example

A shop can analyze the quantity of products sold in individual transactions to understand how small and large quantities are distributed.

### Business Example

A retail business can examine the quartiles of the `Quantity` variable to understand typical order quantities and identify unusually large or small values for further investigation.

### AI/ML Use Case

During data preprocessing, quartiles can help describe numerical features and support the investigation of potential outliers before training a machine learning model.

### Importance in Data Analysis

Quartiles summarize the distribution of a numerical variable and help analysts understand where its values fall.

**Univariate Analysis Rule:** Analyze only one variable at a time. In this example, we analyze `Quantity` alone.

```python
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

quantity_quartiles = quantity.quantile([0.25, 0.50, 0.75])

quantity_quartiles.index = [
    "Q1 (25th Percentile)",
    "Q2 (50th Percentile / Median)",
    "Q3 (75th Percentile)"
]

print("Variable: Quantity")
print("Quartiles of Quantity:")

display(quantity_quartiles.to_frame(name="Value"))
```

## Code Explanation

### 1. Convert the Quantity Column

`pd.to_numeric(df["Quantity"], errors="coerce")` converts the `Quantity` column into numeric values. Any values that cannot be converted become missing values (`NaN`).

### 2. Calculate the Quartiles

`quantity.quantile([0.25, 0.50, 0.75])` calculates the three quartiles:

- `0.25` calculates Q1, the 25th percentile.
- `0.50` calculates Q2, the 50th percentile or median.
- `0.75` calculates Q3, the 75th percentile.

Pandas skips missing values during this calculation.

### 3. Assign Readable Labels

The `index` assignment replaces the default decimal labels with descriptive names so the results are easier to understand.

### 4. Display the Results

`to_frame(name="Value")` converts the quartile results into a table with a column named `Value`. `display()` presents the table clearly.

### How to Interpret the Output

- **Q1:** Approximately 25% of the valid `Quantity` values are at or below this value.
- **Q2:** Approximately 50% of the valid `Quantity` values are at or below this value. This is the median.
- **Q3:** Approximately 75% of the valid `Quantity` values are at or below this value.

Run the code to see the actual quartile values from your dataset. Do not assume the values in advance.

### Impact on the ML Pipeline

Quartiles help data analysts understand numerical feature distributions and investigate potential outliers. This information can guide preprocessing decisions before a dataset is used for machine learning. Quartiles describe `Quantity` alone in this analysis.

# Topic 10: Percentiles

- **90th Percentile:** Approximately 90% of the values are at or below this point.

Percentiles help us understand the position of values within a numerical dataset. They are useful for analyzing distributions and identifying values that may need further investigation.

| Quantity Values |
|---|

# Topic 11: Skewness

## Step 1: Understand the Concept

Skewness is a statistical measure that describes the asymmetry of a numerical variable's distribution.

It helps us understand whether the distribution is balanced or has a longer tail on one side.

There are three common types of skewness:

- **Zero or Near-Zero Skewness:** The distribution is approximately symmetric.
- **Positive Skewness:** The distribution has a longer tail toward the right. A few large values may extend the distribution.
- **Negative Skewness:** The distribution has a longer tail toward the left. A few small values may extend the distribution.

Skewness is useful for understanding the shape of a numerical distribution. A skewness value alone does not explain why the distribution has that shape.

## Step 2: Demonstrate the Concept

### Simple Example

These are examples of possible distribution shapes, not findings from the actual dataset.

### Real-World Example

### Importance in Data Analysis

**Univariate Analysis Rule:** Analyze only one variable at a time. This example examines `Quantity` alone.

# Topic 15: Box Plot

## Step 1: Understand the Concept

A box plot is a visualization used to summarize the distribution of a numerical variable.

It displays important statistical information, including:

- **Minimum and Maximum (Whiskers):** The ends of the whiskers show the range of values within the plot's whisker rule.
- **Q1 (First Quartile):** The 25th percentile.
- **Median (Q2):** The 50th percentile.
- **Q3 (Third Quartile):** The 75th percentile.
- **Potential Outliers:** Individual points beyond the whiskers.

In a standard box plot, the box represents the interquartile range (IQR), calculated as Q3 minus Q1. The whiskers usually extend to the most extreme observations within 1.5 times the IQR from the quartiles. Points beyond the whiskers are displayed separately.

A box plot helps us understand the spread of a numerical variable and identify observations that may need further investigation.

# Topic 21: Pie Chart

## Step 1: Understand the Concept

A pie chart is a circular visualization used to represent the percentage or proportion of categories in a dataset.

The entire circle represents 100% of the observations. Each slice represents the proportion of one category.

The size of each slice depends on the category's frequency relative to the total frequency.

A pie chart is useful when we want to understand how different categories contribute to the whole.

In this analysis, we use the `Country` variable. We display the five most frequent countries separately and combine all remaining countries into an "Other Countries" category to keep the chart readable.

## Step 2: Demonstrate the Concept

### Simple Example

Suppose a dataset contains 100 records:

| Country | Records | Percentage |
|---|---:|---:|
| France | 50 | 50% |
| Germany | 30 | 30% |
| Italy | 20 | 20% |
| Total | 100 | 100% |

In a pie chart, France occupies half of the circle, Germany occupies 30%, and Italy occupies 20%.

### Real-World Example

A retail analyst can use a pie chart to understand the percentage contribution of different countries to the dataset's records.

### Business Example

A business can use the chart to see which country categories account for a larger share of recorded transactions.

### AI/ML Use Case

Pie charts can help with an initial inspection of categorical feature distributions. For detailed comparisons, a bar chart is often easier to interpret.

### Importance in Data Analysis

A pie chart provides a visual understanding of how categories contribute to the total.

**Univariate Analysis Rule:** Analyze only one variable at a time. This pie chart uses `Country` and its record frequencies only.

```python
import matplotlib.pyplot as plt

country_frequency = df["Country"].value_counts()

top_countries = country_frequency.head(5)
other_countries = country_frequency.iloc[5:].sum()

pie_data = top_countries.copy()
pie_data["Other Countries"] = other_countries

plt.figure(figsize=(9, 7))

plt.pie(
    pie_data.values,
    labels=pie_data.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Country Distribution by Percentage")
plt.axis("equal")
plt.show()
```

## Code Explanation

### 1. Calculate Country Frequencies

`df["Country"].value_counts()` counts the non-missing records for each country.

### 2. Select the Top Five Countries

`country_frequency.head(5)` selects the five countries with the highest record counts.

### 3. Combine the Remaining Countries

`country_frequency.iloc[5:].sum()` adds the counts of all countries outside the top five.

`pie_data["Other Countries"] = other_countries` stores that combined count as one additional slice. This changes only the chart data, not the original DataFrame.

### 4. Create the Pie Chart

`plt.pie()` creates the chart using the category counts.

- `pie_data.values` supplies the counts.
- `labels=pie_data.index` displays category names.
- `autopct="%1.1f%%"` displays each slice's percentage to one decimal place.
- `startangle=90` sets the starting angle of the chart.

### 5. Adjust and Display the Chart

`plt.title()` adds a title.

`plt.axis("equal")` keeps the pie chart circular.

`plt.show()` displays the visualization.

### How to Interpret the Output

Each slice represents a country or the combined "Other Countries" category. The percentage printed on each slice shows its share of all non-missing `Country` records.

The percentages should total approximately 100%, allowing for rounding.

### Impact on the ML Pipeline

A pie chart can help identify dominant categories during exploratory data analysis. However, it should not be used alone to decide which categories to remove or combine during machine learning preprocessing.
