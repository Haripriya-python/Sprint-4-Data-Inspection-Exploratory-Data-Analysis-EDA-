## Step 1: Understand the Concept – Histogram

A histogram is a visualization used to understand the distribution of numerical data. It groups numerical values into intervals called bins and shows how many observations fall into each interval.

In the Online Retail II dataset, we can use a histogram to analyze the `Price` column.

### Why Is a Histogram Selected?

A histogram helps us understand:
- Which price ranges occur most frequently.
- Whether prices are concentrated in a particular range.
- Whether the distribution is skewed.
- Whether unusually high or low values may need further investigation.

**AI/ML Application:** Understanding numerical distributions is an important step in Exploratory Data Analysis before preparing data for machine learning.

## Step 2: Demonstrate the Concept – Real-World Example

Imagine an online retail business selling products at different prices.

A histogram can group product prices into intervals such as 0–10, 10–20, 20–30, and so on.

| Price Range | Example Number of Records |
|---|---:|
| 0–10 | 500 |
| 10–20 | 300 |
| 20–30 | 150 |
| Above 30 | 50 |

These numbers are hypothetical and are used only to demonstrate the concept.

If the lower price ranges contain more records, the business may have many low-priced products. The actual distribution must be confirmed using the dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Online Retail II dataset
file_path = "online_retail_II.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="Year 2009-2010"
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
# Select valid, non-missing product prices
price_data = df["Price"].dropna()

# Create a histogram
plt.figure(figsize=(9, 5))
plt.hist(price_data, bins=50, edgecolor="black")

# Add title and labels
plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

# Display the visualization
plt.tight_layout()
plt.show()
```

## Step 3: Interpretation and Insight

### Why This Visualization Was Selected

A histogram was selected because `Price` is a numerical variable. It helps us understand how product prices are distributed across the dataset.

### Interpretation

- **X-axis:** Product price ranges.
- **Y-axis:** Number of records in each price range.
- **Tall bars:** Price ranges containing more records.
- **Short bars:** Price ranges containing fewer records.

### Insight Obtained

The histogram helps identify the price ranges where records are concentrated and whether the distribution has a long tail.

The actual insight must be written after examining the generated graph. For example, if most observations fall in lower price ranges, we can report that lower-priced records are more frequent in this dataset.

### AI/ML Application

This visualization helps us understand the distribution of a numerical variable before modelling. It may also indicate whether further analysis of skewness or potential outliers is needed.

**Important:** Do not remove high or low prices automatically. Investigate them first to determine whether they represent valid retail transactions.

# Topic 2: Box Plot – Price Outliers

## Step 1: Understand the Concept

A **box plot** is a visualization used to understand the distribution of numerical data and identify potential outliers.

In this topic, we use a box plot to analyze the `Price` column in the Online Retail II dataset.

A box plot displays:
- **Median:** The middle value of the data.
- **Quartiles:** Values that divide the data into four parts.
- **Whiskers:** Show the spread of typical data values.
- **Potential outliers:** Values that appear unusually far from the rest of the data.

### Why Did We Select a Box Plot?

Product prices may contain unusually high or low values. A box plot helps us identify these values and understand how widely prices vary.

**Objective:** Visualize the distribution of product prices and identify potential outliers without automatically removing them.

## Step 2: Demonstrate the Concept

Imagine a shop selling products at different prices.

| Product | Price |
|---|---:|
| Notebook | ₹20 |
| Pen | ₹30 |
| Water bottle | ₹100 |
| Bag | ₹500 |
| Premium product | ₹5,000 |

Most products may have relatively low prices, while a few products may be much more expensive.

A box plot helps us see the middle range of prices and identify values that fall unusually far from the rest.

**Business example:** An online retailer can investigate unusually high product prices to check whether they represent premium products, data-entry mistakes, or other valid cases.

**AI/ML use:** Identifying potential outliers helps us understand the data before preparing it for machine learning. An outlier should be investigated before deciding whether to keep or remove it.

```python

# Select valid, non-missing product prices
price_data = df["Price"].dropna()

# Create a box plot
plt.figure(figsize=(9, 4))
plt.boxplot(price_data, vert=False)

# Add title and axis label
plt.title("Box Plot of Product Prices")
plt.xlabel("Price")

# Display the visualization
plt.tight_layout()
plt.show()
```

# Topic 3: Bar Chart – Average Price by Country

## Step 1: Understand the Concept

A **bar chart** is a visualization used to compare values across different categories.

In this topic, we use a bar chart to compare the average product price across countries in the Online Retail II dataset.

- **X-axis:** Country names.
- **Y-axis:** Average product price.

### Why Did We Select a Bar Chart?

The dataset contains transactions from different countries. A bar chart makes it easier to compare average prices between countries and identify differences across categories.

**Objective:** Compare average product prices across a selection of countries and understand how prices vary by country.

## Step 2: Demonstrate the Concept

Imagine an online retailer selling products to customers in different countries.

| Country | Example Average Price |
|---|---:|
| Country A | ₹100 |
| Country B | ₹150 |
| Country C | ₹250 |

These are hypothetical values used only to demonstrate the concept.

A bar chart helps compare the average price for each country. Taller bars represent higher average prices, while shorter bars represent lower average prices.

**Business example:** A retailer can compare average transaction prices across markets and investigate differences in product mix or pricing.

**AI/ML and Data Analytics use:** Bar charts help analysts compare numerical measurements across categories before building reports or preparing data for analysis.

```python

# Remove rows with missing country or price values
country_price_data = df[["Country", "Price"]].dropna()

# Calculate average price for each country
avg_price_by_country = (
    country_price_data.groupby("Country")["Price"]
    .mean()
    .sort_values(ascending=False)
)

# Select the 10 countries with the highest average prices
top_10_countries = avg_price_by_country.head(10)

# Create a bar chart
plt.figure(figsize=(12, 6))
plt.bar(top_10_countries.index, top_10_countries.values)

# Add title and labels
plt.title("Top 10 Countries by Average Product Price")
plt.xlabel("Country")
plt.ylabel("Average Price")
plt.xticks(rotation=45, ha="right")

# Display the visualization
plt.tight_layout()
plt.show()
```

## Interpretation and Analysis

### 1. Why Was This Visualization Selected?

A bar chart was selected because country is a categorical variable and average price is a numerical measurement. This chart allows us to compare the measurements across categories.

### 2. How to Interpret the Visualization

- **X-axis:** Displays the selected country names.
- **Y-axis:** Displays the average product price for each country.
- **Bar height:** Represents the average price calculated from the available records.

The countries are displayed in descending order of average price.

### 3. Insight Obtained

After running the chart, identify which countries have higher or lower average product prices among the selected ten countries.

**Important:** These are averages across transaction records, not necessarily the prices of comparable products. Differences may reflect the product mix or unusual transactions. The chart alone does not establish why the differences occur.

# Topic 6: Line Plot – Sales Trend Over Time

## Step 1: Understand the Concept

A **line plot** is a visualization used to show how a numerical value changes over time.

In this topic, we use the `InvoiceDate`, `Quantity`, and `Price` columns from the Online Retail II dataset to explore daily sales value over time.

We calculate an approximate sales value for each transaction row by multiplying `Quantity` by `Price`, then aggregate it by day.

- **X-axis:** Invoice date.
- **Y-axis:** Approximate daily sales value.

### Why Did We Select a Line Plot?

A line plot makes it easier to observe changes, trends, peaks, and drops over time.

**Objective:** Explore how the recorded daily sales value changes throughout the available period.

## Step 2: Demonstrate the Concept

Imagine an online store recording its sales value each day.

| Date | Example Daily Sales Value |
|---|---:|
| 1 January | ₹5,000 |
| 2 January | ₹7,000 |
| 3 January | ₹4,000 |
| 4 January | ₹9,000 |

These values are hypothetical and are used only to explain the concept.

A line plot connects the daily values in date order. An upward movement indicates an increase, while a downward movement indicates a decrease.

**Business example:** A retailer can use daily sales trends to investigate busy periods, changes in demand, or unusual drops.

**AI/ML and Data Analytics use:** Time-based visualizations help analysts understand historical patterns before considering forecasting or other time-series analysis.

```python

# Select required columns and remove missing values
sales_data = df[["InvoiceDate", "Quantity", "Price"]].dropna().copy()

# Convert invoice dates to datetime format
sales_data["InvoiceDate"] = pd.to_datetime(
    sales_data["InvoiceDate"],
    errors="coerce"
)

# Remove rows with invalid dates
sales_data = sales_data.dropna(subset=["InvoiceDate"])

# Calculate approximate transaction-row sales value
sales_data["SalesValue"] = (
    sales_data["Quantity"] * sales_data["Price"]
)

# Aggregate sales value by day
daily_sales = (
    sales_data.groupby(sales_data["InvoiceDate"].dt.date)["SalesValue"]
    .sum()
    .sort_index()
)

# Create a line plot
plt.figure(figsize=(12, 6))
plt.plot(daily_sales.index, daily_sales.values)

# Add title and labels
plt.title("Daily Sales Value Over Time")
plt.xlabel("Invoice Date")
plt.ylabel("Approximate Sales Value")
plt.xticks(rotation=45)

# Display the visualization
plt.tight_layout()
plt.show()
```

## Interpretation and Analysis

### 1. Why Was This Visualization Selected?

A line plot was selected because the analysis examines changes in a numerical value over time. Connecting daily values helps reveal trends and fluctuations.

### 2. How to Interpret the Visualization

- **X-axis:** Displays invoice dates in chronological order.
- **Y-axis:** Displays the sum of `Quantity × Price` for each date.
- **Upward sections:** Indicate an increase in recorded daily sales value.
- **Downward sections:** Indicate a decrease in recorded daily sales value.
- **Peaks and drops:** May identify unusually active or quiet days that need further investigation.

### 3. Insight Obtained

After running the plot, observe whether daily sales values fluctuate, whether any dates show peaks or drops, and whether a broader trend is visible.

**Important:** This is an approximate transaction-row sales measure, not necessarily official net revenue. Negative quantities, cancellations, returns, and other transaction conditions can affect the totals. The chart covers the dates available in the selected dataset sheet.

In this topic, we use the `Country` and `Price` columns from the Online Retail II dataset to compare product price distributions across selected countries.

- **X-axis:** Country names.
- **Y-axis:** Product prices.

### Why Did We Select a Violin Plot?

A violin plot combines distribution shape with information about where values are concentrated. It helps us compare price variation across countries.

**Objective:** Examine how product prices are distributed across selected countries and identify differences in their spread and concentration.

## Step 2: Demonstrate the Concept

Imagine an online store selling products in three countries.

| Country | Example Product Prices |
|---|---|
| Country A | 10, 15, 20, 25, 30 |
| Country B | 20, 30, 40, 50, 60 |
| Country C | 10, 10, 15, 15, 100 |

These values are hypothetical.

A violin plot helps us compare the distribution of prices across categories. Wider sections indicate a higher concentration of values around those price levels.

**Business example:** A retailer can compare price variation across markets and investigate whether the observed differences relate to product selection or unusual prices.

**AI/ML and Data Analytics use:** Distribution plots help analysts understand feature patterns and identify differences between groups before further analysis.

```python

# Select country and price columns
violin_data = df[["Country", "Price"]].dropna().copy()

# Keep positive prices for clearer distribution analysis
violin_data = violin_data[violin_data["Price"] > 0]

# Select the 5 countries with the most transaction records
selected_countries = (
    df["Country"].value_counts().head(5).index
)

violin_data = violin_data[
    violin_data["Country"].isin(selected_countries)
]

# Limit extreme price values using the 99th percentile for readability
price_limit = violin_data["Price"].quantile(0.99)
violin_data = violin_data[
    violin_data["Price"] <= price_limit
]

# Create a violin plot
plt.figure(figsize=(12, 6))
sns.violinplot(
    data=violin_data,
    x="Country",
    y="Price"
)

# Add title and labels
plt.title("Product Price Distribution by Country")
plt.xlabel("Country")
plt.ylabel("Price")
plt.xticks(rotation=45, ha="right")

# Display the visualization
plt.tight_layout()
plt.show()
```

## Interpretation and Analysis

### 2. How to Interpret the Visualization

- **Different shapes:** Suggest that price distributions vary across countries.

# Topic 9: Heatmap – Correlation Between Numerical Features

## Step 1: Understand the Concept

A **heatmap** is a visualization that uses colors to represent numerical values in a table-like format.

In this topic, we use a correlation matrix to examine the relationships between numerical features in the Online Retail II dataset.

We will analyze:
- `Quantity`
- `Price`
- `Customer ID`

**Correlation** measures the strength and direction of a linear relationship between two numerical variables. Its value ranges from -1 to +1.

- **+1:** Perfect positive linear correlation.
- **0:** No linear correlation.
- **-1:** Perfect negative linear correlation.

### Why Did We Select a Heatmap?

A heatmap makes it easier to compare several correlation values at once and identify possible linear relationships between numerical features.

**Objective:** Visualize the correlation matrix and examine the linear relationships between selected numerical columns.

## Step 2: Demonstrate the Concept

Imagine a retailer examining the relationship between product quantity and price.

| Feature Pair | Example Correlation |
|---|---:|
| Quantity and Price | -0.60 |
| Quantity and Sales Value | 0.75 |
| Price and Sales Value | 0.30 |

These values are hypothetical and are only used to demonstrate the concept.

A heatmap displays each correlation value using a color. The color intensity helps us compare stronger and weaker linear relationships.

**Business example:** A retailer can investigate whether numerical transaction features tend to change together.

**AI/ML use:** Correlation analysis can help identify relationships between numerical features before model development. It can also help identify potentially redundant features.

**Important:** Correlation does not prove causation. A correlation near zero means little or no linear relationship, but a nonlinear relationship may still exist.

```python

# Select numerical columns for correlation analysis
numeric_data = df[["Quantity", "Price", "Customer ID"]].copy()

# Convert selected columns to numeric values where possible
numeric_data = numeric_data.apply(
    pd.to_numeric,
    errors="coerce"
)

# Calculate the correlation matrix
correlation_matrix = numeric_data.corr()

# Create a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

# Add title
plt.title("Correlation Heatmap of Numerical Features")

# Display the visualization
plt.tight_layout()
plt.show()
```

## Interpretation and Analysis

### 1. Why Was This Visualization Selected?

A heatmap was selected because it displays multiple correlation coefficients in one figure, making comparisons easier.

### 2. How to Interpret the Visualization

- **Values near +1:** Indicate a strong positive linear relationship.
- **Values near -1:** Indicate a strong negative linear relationship.
- **Values near 0:** Indicate a weak or absent linear relationship.
- **Diagonal values:** Equal 1 because each variable is perfectly correlated with itself.

### 3. Insight Obtained

After running the heatmap, examine the actual correlation values for `Quantity`, `Price`, and `Customer ID`. Identify which feature pairs have stronger positive or negative correlations and which have weaker linear relationships.

**Important:** `Customer ID` is an identifier, not a meaningful continuous measurement. Correlations involving it may be misleading and should not be interpreted as business relationships. Missing values are excluded pairwise when calculating correlations, and correlation alone does not establish cause and effect.

# Topic 10: Distribution Plot – Price Distribution

## Step 1: Understand the Concept

A **distribution plot** helps us understand how numerical values are spread across a dataset. It can show the frequency of values and, when a density curve is included, indicate where values are concentrated.

In this topic, we analyze the `Price` column from the Online Retail II dataset.

We will examine:
- The range of product prices.
- Where prices are concentrated.
- Whether the distribution is symmetric or skewed.
- Whether unusually high prices are present.

### Why Did We Select a Distribution Plot?

A distribution plot helps us understand the shape of numerical data. This is useful for identifying skewness and deciding whether further investigation is needed.

**Objective:** Examine the distribution of product prices and understand how frequently different price ranges occur.

## Step 2: Demonstrate the Concept

Imagine an online shop selling products at different prices.

| Product Price | Example Number of Products |
|---|---:|
| ₹0–₹50 | 100 |

**Business example:** A retailer can understand its product price distribution and investigate whether a small number of expensive products affect the overall pattern.

**AI/ML and Data Analytics use:** Understanding feature distributions helps analysts identify skewness, unusual values, and possible preprocessing requirements before modeling.

```python

# Select valid, positive product prices
price_data = df["Price"].dropna()
price_data = price_data[price_data > 0]

# Limit the display to the 99th percentile for readability
price_limit = price_data.quantile(0.99)
price_data_display = price_data[price_data <= price_limit]

# Create a distribution plot
plt.figure(figsize=(9, 5))
sns.histplot(
    price_data_display,
    bins=50,
    kde=True
)

# Add title and labels
plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

# Display the visualization
plt.tight_layout()
plt.show()
```

- **Histogram bars:** Show how many observations fall within each price interval.
- **Density curve:** Shows the estimated shape of the distribution.
- **Long tail on the right:** May indicate a right-skewed distribution, where a small number of prices are much higher than most prices.
