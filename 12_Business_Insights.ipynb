
# Step 1: Understand the Concept

## Identifying Important Patterns

Important patterns are repeated or noticeable behaviors in business data that help us understand business performance.

In an online retail dataset, patterns can be identified by analyzing sales value, quantity sold, number of transactions, and customer activity.

Identifying these patterns helps businesses understand which areas perform well and which areas may need improvement.

### Objective
- Identify important patterns in retail transaction data.
- Summarize sales value and quantity sold by country.
- Use the results to support meaningful business observations.



# Step 2: Demonstrate the Concept

## Simple Business Example

An online retailer sells products to customers in different countries.

By grouping transaction data by country, the business can compare the recorded sales value and quantity sold in each country.

| Country | Recorded Sales Value | Quantity Sold |
|---|---:|---:|
| Country A | 50,000 | 2,000 |
| Country B | 30,000 | 1,500 |
| Country C | 10,000 | 800 |

These values are illustrative examples, not actual results from the dataset.

### Business Understanding

- Country A has the highest recorded sales value in this example.
- Country C has the lowest recorded sales value.
- These differences may help identify important business patterns.

The actual dataset must be analyzed before drawing conclusions.


```python

import pandas as pd
import numpy as np

# Load the retail dataset
file_path = "online_retail_II.xlsx"
df = pd.read_excel(file_path, sheet_name="Year 2009-2010")

# Create a copy for analysis
retail_df = df.copy()

# Convert required columns to numeric values
retail_df["Quantity"] = pd.to_numeric(
    retail_df["Quantity"], errors="coerce"
)
retail_df["Price"] = pd.to_numeric(
    retail_df["Price"], errors="coerce"
)

# Calculate recorded transaction-row sales value
retail_df["SalesValue"] = (
    retail_df["Quantity"] * retail_df["Price"]
)

# Remove rows missing country or required numeric values
pattern_df = retail_df.dropna(
    subset=["Country", "Quantity", "Price", "SalesValue"]
)

# Summarize sales value and quantity by country
country_patterns = (
    pattern_df.groupby("Country")
    .agg(
        Total_Sales_Value=("SalesValue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Transaction_Rows=("Invoice", "count")
    )
    .sort_values("Total_Sales_Value", ascending=False)
)

print("Top 10 Countries by Recorded Sales Value:")
display(country_patterns.head(10).round(2))

print("\nCountries with the Lowest Recorded Sales Value:")
display(country_patterns.tail(10).sort_values("Total_Sales_Value").round(2))
```


# Interpretation and Business Insight

## Code Explanation

- `pd.to_numeric()` converts Quantity and Price into numeric values where possible.
- `SalesValue` calculates the recorded transaction-row value by multiplying Quantity by Price.
- `groupby("Country")` groups the transaction rows by country.
- `agg()` summarizes sales value, quantity, and transaction-row count.
- `sort_values()` arranges countries by recorded sales value.
- `head(10)` displays the ten countries with the highest recorded sales value.
- `tail(10)` identifies the countries with the lowest recorded sales value.

## Business Observation

Use the actual output to identify:
- Countries contributing the highest recorded sales value.
- Countries contributing the lowest recorded sales value.
- Whether high sales value is accompanied by high quantity sold.
- Countries that may need further investigation.

## Business Insight

Countries with high recorded sales value may represent important markets for the business. Countries with lower recorded sales value may require further analysis to understand customer demand, transaction activity, and potential growth opportunities.

These are possible interpretations. Final conclusions must be based on the actual results.

## Important Note

SalesValue is calculated as Quantity multiplied by Price. Negative quantities, including returns or adjustments, can affect the total. This is a transaction-row sales-value analysis, not necessarily the company's final net revenue.

## Identifying Trends

A trend is a general pattern of change in business data over time.

In retail data, trends help us understand how recorded sales value and quantity sold change across different dates or months.

Analyzing monthly trends helps businesses identify periods of higher or lower sales activity.

### Objective
- Analyze monthly recorded sales value.
- Identify months with higher or lower recorded sales value.
- Understand how time-based patterns can support business decisions.



# Step 2: Demonstrate the Concept

## Simple Business Example

An online retailer compares its recorded sales value across three months.

| Month | Recorded Sales Value |
|---|---:|
| January | 20,000 |
| February | 28,000 |
| March | 35,000 |

These values are illustrative examples, not actual dataset results.

### Business Understanding

- Recorded sales value increases from January to March.
- The business can investigate possible reasons for this increase.
- Monthly comparisons can help identify changes in business activity.

Actual trends must be identified using the retail dataset.


```python

# Convert InvoiceDate to datetime
retail_df["InvoiceDate"] = pd.to_datetime(
    retail_df["InvoiceDate"], errors="coerce"
)

# Keep rows with valid dates and calculated sales values
trend_df = retail_df.dropna(
    subset=["InvoiceDate", "SalesValue"]
).copy()

# Extract the month for grouping
trend_df["Month"] = (
    trend_df["InvoiceDate"].dt.to_period("M")
)

# Calculate monthly sales value and quantity
monthly_trends = (
    trend_df.groupby("Month")
    .agg(
        Total_Sales_Value=("SalesValue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Transaction_Rows=("Invoice", "count")
    )
    .sort_index()
)

print("Monthly Retail Trends:")
display(monthly_trends.round(2))

# Identify months with the highest and lowest recorded sales value
if not monthly_trends.empty:
    highest_month = monthly_trends["Total_Sales_Value"].idxmax()
    lowest_month = monthly_trends["Total_Sales_Value"].idxmin()

    print("\nMonth with Highest Recorded Sales Value:", highest_month)
    print(
        "Recorded Sales Value:",
        round(monthly_trends.loc[highest_month, "Total_Sales_Value"], 2)
    )

    print("\nMonth with Lowest Recorded Sales Value:", lowest_month)
    print(
        "Recorded Sales Value:",
        round(monthly_trends.loc[lowest_month, "Total_Sales_Value"], 2)
    )
else:
    print("No valid monthly data available for analysis.")
```


# Interpretation and Business Insight

## Code Explanation

- `pd.to_datetime()` converts InvoiceDate into a datetime format.
- `.dt.to_period("M")` groups dates by month.
- `groupby("Month")` organizes transaction rows by month.
- `agg()` calculates monthly recorded sales value, quantity, and transaction-row count.
- `idxmax()` identifies the month with the highest recorded sales value.
- `idxmin()` identifies the month with the lowest recorded sales value.

## Business Observation

Use the actual output to identify:
- Months with higher recorded sales value.
- Months with lower recorded sales value.
- Whether monthly quantity sold changes along with recorded sales value.
- Any noticeable increases or decreases across the available months.

## Business Insight

Monthly trends can help a retailer understand changes in sales activity and investigate periods that perform differently.

If a month has higher recorded sales value, the business can investigate product demand and customer activity during that period. If a month has lower recorded sales value, it can examine possible reasons before deciding on any action.

Conclusions must be based on the actual dataset results.

## Important Note

This analysis uses Quantity multiplied by Price as recorded transaction-row sales value. Negative quantities, including returns or adjustments, can affect monthly totals.

## Identifying Relationships

Relationships describe how two or more variables behave in relation to each other.

In retail data, Quantity and Price can be analyzed to understand how they relate to recorded Sales Value.

Understanding relationships helps businesses identify useful patterns and investigate factors associated with business performance.

### Objective
- Analyze the relationship between Quantity and Price.
- Examine how these variables relate to recorded Sales Value.
- Identify observations that may help business analysis.



# Step 2: Demonstrate the Concept

## Simple Business Example

An online retailer records the following transactions.

| Transaction | Quantity | Price per Unit | Recorded Sales Value |
|---|---:|---:|---:|
| A | 2 | 100 | 200 |
| B | 4 | 100 | 400 |
| C | 3 | 200 | 600 |

These values are illustrative examples, not actual dataset results.

### Business Understanding

- Recorded Sales Value depends on both Quantity and Price.
- A transaction with a higher quantity does not always have a higher sales value if its unit price is lower.
- Relationships help businesses understand how different variables contribute to recorded transaction values.


```python

# Select the variables for relationship analysis
relationship_df = retail_df[
    ["Quantity", "Price", "SalesValue"]
].dropna()

# Remove infinite values
relationship_df = relationship_df.replace(
    [np.inf, -np.inf], np.nan
).dropna()

# Calculate the correlation between the variables
correlation_matrix = relationship_df.corr()

print("Correlation Between Quantity, Price, and Sales Value:")
display(correlation_matrix.round(3))

# Compare Quantity and Price with Sales Value
print("\nQuantity and Sales Value Correlation:")
print(round(correlation_matrix.loc["Quantity", "SalesValue"], 3))

print("\nPrice and Sales Value Correlation:")
print(round(correlation_matrix.loc["Price", "SalesValue"], 3))
```


# Interpretation and Business Insight

## Code Explanation

- The selected columns contain Quantity, Price, and SalesValue.
- `dropna()` removes rows with missing values.
- `replace()` converts infinite values into missing values before removing them.
- `corr()` calculates Pearson correlation coefficients between the numeric variables.
- Correlation values range from -1 to +1.

## Business Observation

Use the actual correlation results to identify:
- Whether Quantity and Sales Value have a positive or negative linear relationship.
- Whether Price and Sales Value have a positive or negative linear relationship.
- Which pair of variables has a stronger linear relationship in the results.

## Business Insight

Understanding relationships between retail variables can help businesses investigate patterns in transaction values and product activity.

However, correlation does not prove causation. Sales Value is calculated using Quantity multiplied by Price, so relationships involving Sales Value must be interpreted with this calculation in mind.

## Important Note

The dataset contains returns and negative quantities. These records can affect correlation results, so the results should be interpreted in the context of the original transaction data.

## Identifying Anomalies

Anomalies are observations that differ noticeably from the usual patterns in a dataset.

In retail data, examples may include unusually large quantities, unusually high prices, or negative quantities that may represent returns or adjustments.

Identifying anomalies helps businesses investigate unusual transactions and possible data quality problems.

### Objective
- Identify unusual values in Quantity and Price.
- Examine transactions with negative quantities.
- Understand why unusual records may need further investigation.



# Step 2: Demonstrate the Concept

## Simple Business Example

An online retailer records the following transactions.

| Transaction | Quantity | Price |
|---|---:|---:|
| A | 2 | 100 |
| B | 3 | 120 |
| C | 500 | 100 |
| D | -2 | 100 |

These values are illustrative examples, not actual dataset results.

### Business Understanding

- Transaction C has an unusually high quantity compared with the other examples.
- Transaction D has a negative quantity, which may indicate a return or adjustment.
- These records should be investigated before deciding whether they are valid or incorrect.

An unusual value is not automatically an error.


```python

# Select relevant columns and remove missing values
anomaly_df = retail_df[
    ["Invoice", "Description", "Quantity", "Price", "InvoiceDate", "Country"]
].dropna(subset=["Quantity", "Price"]).copy()

# Calculate IQR limits for Quantity
q1_quantity = anomaly_df["Quantity"].quantile(0.25)
q3_quantity = anomaly_df["Quantity"].quantile(0.75)
iqr_quantity = q3_quantity - q1_quantity

lower_limit = q1_quantity - 1.5 * iqr_quantity
upper_limit = q3_quantity + 1.5 * iqr_quantity

# Identify unusual Quantity values using the IQR rule
quantity_anomalies = anomaly_df[
    (anomaly_df["Quantity"] < lower_limit) |
    (anomaly_df["Quantity"] > upper_limit)
]

# Identify negative quantities
negative_quantity_rows = anomaly_df[
    anomaly_df["Quantity"] < 0
]

print("Quantity IQR Lower Limit:", lower_limit)
print("Quantity IQR Upper Limit:", upper_limit)

print("\nNumber of Unusual Quantity Records:", len(quantity_anomalies))
display(quantity_anomalies.head(10))

print("\nNumber of Records with Negative Quantity:", len(negative_quantity_rows))
display(negative_quantity_rows.head(10))
```


# Interpretation and Business Insight

## Code Explanation

- `quantile(0.25)` calculates the first quartile (Q1).
- `quantile(0.75)` calculates the third quartile (Q3).
- The IQR is the difference between Q3 and Q1.
- The IQR rule identifies Quantity values below the lower limit or above the upper limit.
- The second filter identifies records with negative quantities.

## Business Observation

Use the actual output to identify:
- Whether unusually large or small quantities are present.
- How many records have negative quantities.
- Whether unusual records appear to represent returns, adjustments, or transactions requiring further review.

## Business Insight

Unusually large quantities may indicate bulk purchases or records that require verification. Negative quantities may represent returns or adjustments and can affect recorded sales totals.

Investigating these records can help improve data quality and support more reliable business reporting.

## Important Note

The IQR rule flags unusual values; it does not prove that they are errors. Do not automatically remove these records. Check the transaction details and business context before making a decision.

## Identifying Potential Business Problems

Business problems are situations in the data that may require further investigation or improvement.

In retail data, possible problems may include low sales activity in certain markets, frequent negative-quantity transactions, missing product descriptions, or incomplete customer information.

EDA helps identify warning signs, but additional investigation is needed before confirming a business problem.

### Objective
- Examine missing values in important columns.
- Identify countries with lower recorded sales values.
- Identify data quality issues that may affect business analysis.



# Step 2: Demonstrate the Concept

## Simple Business Example

An online retailer reviews its transaction data.

| Observation | Possible Business Concern |
|---|---|
| Some transactions have missing customer IDs | Customer-level analysis may be limited |
| Some products have missing descriptions | Product identification may be difficult |
| Some countries have lower recorded sales values | Market performance may need investigation |

These are illustrative examples, not confirmed findings from the dataset.

### Business Understanding

- Missing values may reduce the usefulness of certain analyses.
- Lower recorded sales value does not automatically mean poor performance.
- Businesses should investigate the cause before deciding on corrective actions.


```python

# Check missing values in important columns
important_columns = [
    "Customer ID", "Description", "Invoice",
    "Quantity", "Price", "Country"
]

missing_summary = pd.DataFrame({
    "Missing_Count": retail_df[important_columns].isna().sum(),
    "Missing_Percentage": (
        retail_df[important_columns].isna().mean() * 100
    )
})

print("Missing Values in Important Columns:")
display(missing_summary.round(2))

# Calculate recorded sales value by country
business_problem_df = retail_df.dropna(
    subset=["Country", "SalesValue"]
).copy()

country_sales = (
    business_problem_df.groupby("Country")
    .agg(
        Total_Sales_Value=("SalesValue", "sum"),
        Transaction_Rows=("Invoice", "count")
    )
    .sort_values("Total_Sales_Value")
)

print("\nCountries with the Lowest Recorded Sales Value:")
display(country_sales.head(10).round(2))

# Check negative-quantity transactions
negative_quantity_count = (
    retail_df["Quantity"] < 0
).sum()

print("\nRecords with Negative Quantity:", negative_quantity_count)
```


# Interpretation and Business Insight

## Code Explanation

- `isna().sum()` counts missing values in each selected column.
- `isna().mean() * 100` calculates the percentage of missing values.
- `groupby("Country")` summarizes transaction data by country.
- `sort_values()` arranges countries by recorded sales value.
- The negative-quantity condition counts records that may include returns or adjustments.

## Business Observation

Use the actual output to identify:
- Columns with a high number or percentage of missing values.
- Countries with lower recorded sales value.
- The number of transactions containing negative quantities.

## Business Insight

Missing customer IDs may limit customer behavior analysis, while missing product descriptions may make product-level reporting less reliable.

Countries with lower recorded sales value may need further investigation into transaction volume, demand, and market activity. Negative quantities should be checked to understand their effect on recorded sales totals.

## Important Note

These observations indicate areas for investigation, not confirmed business problems. Lower sales value alone does not prove that a country or market is underperforming.

## Identifying Interesting Segments

Segmentation is the process of dividing business data into meaningful groups based on shared characteristics.

In retail data, transactions can be grouped by country, product, or customer to understand differences in business activity.

Identifying interesting segments helps businesses discover which groups contribute differently to recorded sales value and transaction activity.

### Objective
- Identify countries with different levels of recorded sales value.
- Compare transaction activity across countries.
- Identify segments that may require further business analysis.



# Step 2: Demonstrate the Concept

## Simple Business Example

An online retailer compares recorded sales value across different customer markets.

| Segment | Recorded Sales Value |
|---|---:|
| Segment A | 50,000 |
| Segment B | 30,000 |
| Segment C | 10,000 |

These values are illustrative examples, not actual dataset results.

### Business Understanding

- Segment A contributes the highest recorded sales value in this example.
- Segment C contributes the lowest recorded sales value.
- Comparing segments helps businesses identify differences in market activity.

Actual segments must be analyzed using the retail dataset.


```python

# Prepare data for country-level segmentation
segment_df = retail_df.dropna(
    subset=["Country", "SalesValue"]
).copy()

# Summarize recorded sales value and transaction activity
country_segments = (
    segment_df.groupby("Country")
    .agg(
        Total_Sales_Value=("SalesValue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Transaction_Rows=("Invoice", "count"),
        Unique_Invoices=("Invoice", "nunique")
    )
    .sort_values("Total_Sales_Value", ascending=False)
)

print("Top 10 Country Segments by Recorded Sales Value:")
display(country_segments.head(10).round(2))

print("\nBottom 10 Country Segments by Recorded Sales Value:")
display(
    country_segments.tail(10)
    .sort_values("Total_Sales_Value")
    .round(2)
)
```


# Interpretation and Business Insight

## Code Explanation

- `groupby("Country")` divides the data into country-based segments.
- `Total_Sales_Value` summarizes recorded sales value for each country.
- `Total_Quantity` summarizes the quantities recorded for each country.
- `Transaction_Rows` counts transaction rows.
- `Unique_Invoices` counts distinct invoice identifiers.
- `head(10)` and `tail(10)` show the highest and lowest segments by recorded sales value.

## Business Observation

Use the actual output to identify:
- Country segments with higher recorded sales value.
- Country segments with lower recorded sales value.
- Differences between transaction-row counts and unique invoice counts.
- Segments that may need further investigation.

## Business Insight

Country-level segmentation can help a retailer understand how recorded sales value and transaction activity differ across markets.

Higher recorded sales value may indicate an important market, while lower recorded sales value may provide an opportunity for further investigation. Transaction activity should also be considered before drawing conclusions.

## Important Note

Transaction rows are not the same as unique invoices. Sales values can also be affected by returns and adjustments, so segment comparisons should be interpreted carefully.

## Identifying Potential Business Opportunities

Business opportunities are areas where a company may improve performance, serve customers better, or increase sales.

### Objective

- Identify products with lower recorded sales value.
- Find potential areas for further business investigation.



# Step 2: Demonstrate the Concept

## Simple Business Example

An online retailer compares the recorded sales value of different products.

| Product | Recorded Sales Value |
|---|---:|
| Product A | 40,000 |
| Product B | 25,000 |
| Product C | 8,000 |

These values are illustrative examples, not actual dataset results.

### Business Understanding

- Product A has the highest recorded sales value in this example.
- Product C has the lowest recorded sales value.
- The business may investigate product demand, pricing, and customer preferences to understand these differences.

Lower recorded sales value does not automatically mean that a product should be discontinued.


```python

# Prepare data for product-level analysis
product_df = retail_df.dropna(
    subset=["StockCode", "Description", "SalesValue"]
).copy()

# Summarize recorded sales value and quantity by product
product_opportunities = (
    product_df.groupby(["StockCode", "Description"])
    .agg(
        Total_Sales_Value=("SalesValue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Transaction_Rows=("Invoice", "count")
    )
    .sort_values("Total_Sales_Value", ascending=False)
)

print("Top 10 Products by Recorded Sales Value:")
display(product_opportunities.head(10).round(2))

print("\nBottom 10 Products by Recorded Sales Value:")
display(
    product_opportunities.tail(10)
    .sort_values("Total_Sales_Value")
    .round(2)
)
```


# Interpretation and Business Insight

## Code Explanation

- `groupby(["StockCode", "Description"])` groups transaction records by product.
- `Total_Sales_Value` summarizes recorded sales value for each product.
- `Total_Quantity` summarizes the recorded quantity for each product.
- `Transaction_Rows` counts the transaction rows associated with each product.
- `sort_values()` arranges products by recorded sales value.

## Business Observation

Use the actual output to identify:
- Products with higher recorded sales value.
- Products with lower recorded sales value.
- Differences in quantity and transaction activity across products.

## Business Insight

Products with higher recorded sales value may deserve closer analysis to understand their contribution to the business. Products with lower recorded sales value may provide opportunities to investigate demand, pricing, availability, and customer preferences.

These findings can guide further investigation, but they do not independently prove which products should receive more investment or be discontinued.

## Important Note

Recorded sales value is calculated as Quantity multiplied by Price. Returns, adjustments, and unusual quantities can affect product totals, so review the transaction details before making business decisions.
