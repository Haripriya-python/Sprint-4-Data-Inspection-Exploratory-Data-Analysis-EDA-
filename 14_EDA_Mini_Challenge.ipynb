# Topic 4: Identify Data Types
## Step 1 – Understand the Concept

Identifying data types means checking how each feature is stored in a dataset.

## Step 2 – Demonstrate the Concept

### 1. Simple Example

Consider these values:

| Feature | Example | Possible Data Type |
|---|---|---|
| Quantity | 5 | Integer |
| Price | 25.50 | Float |
| Product | Notebook | String |
| InvoiceDate | 2026-10-22 | Datetime |

Each data type supports different operations.

### 2. Real-World Example

In an online retail dataset, Quantity and Price are numerical features, Description is text, and InvoiceDate represents transaction timing.

The actual stored data types must be checked rather than assumed.

### 3. Business Example

A retail analyst needs numerical data types to calculate quantities and prices and a valid date type to summarize transactions by month.

If a date is stored as text, date-based calculations may not work as expected.

### 4. AI/ML Use Case

Machine learning pipelines require features in suitable formats. Numerical features may be scaled, categorical features may be encoded, and date features may be transformed into useful time-based variables.

Identifiers such as Invoice and StockCode should not automatically be treated as numerical measurements.

### 5. Why This Concept Is Important in Data Analysis

Checking data types helps prevent calculation errors, supports accurate visualizations, and identifies columns that may require correction during preprocessing.


```python

# Identify the data types of all dataset features

dtype_summary = pd.DataFrame({
    "Feature": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Non_Null_Count": df.notna().sum().values,
    "Unique_Values": df.nunique(dropna=True).values
})

print("Feature Data Type Summary:")
display(dtype_summary)

# Group features by their current data types
print("\nNumerical Features:")
print(
    df.select_dtypes(include=["number"]).columns.tolist()
)

print("\nNon-Numerical Features:")
print(
    df.select_dtypes(exclude=["number"]).columns.tolist()
)

# Check whether InvoiceDate can be interpreted as a date
converted_dates = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

print("\nCurrent InvoiceDate Data Type:")
print(df["InvoiceDate"].dtype)

print(
    "Missing or Invalid Dates After Conversion Check:",
    converted_dates.isna().sum()
)
```


## Step 4 – Code Explanation and Interpretation

### 1. What the Code Does

The code displays each feature's current data type, non-missing value count, and number of unique values. It also separates numerical and non-numerical features and checks whether InvoiceDate can be interpreted as a date.

### 2. Why These Methods Were Used

- `df.dtypes` identifies the stored data type of each feature.
- `df.notna().sum()` counts non-missing values.
- `df.nunique()` counts distinct non-missing values.
- `select_dtypes()` groups features based on their current types.
- `pd.to_datetime()` checks date conversion.
- `errors="coerce"` marks unconvertible values as `NaT` in the converted result.

The date check does not modify the original InvoiceDate column.

### 3. What the Output Means

The summary shows how each feature is stored. The date check reports how many values are missing or invalid after conversion.

A numerical identifier may still represent a category rather than a measurable quantity. The output must be interpreted using the meaning of each feature.

### 4. Insights That Can Be Obtained

# Topic 5: Identify Missing Values
## Step 1 – Understand the Concept

Missing values are data entries that do not contain an available value.

They may occur because information was not collected, was unavailable, or was not recorded correctly.

Identifying missing values involves:
- Counting missing values in each feature.
- Calculating the percentage of missing values.
- Identifying features affected by missing data.
- Understanding how missing data may affect analysis.

Missing values should be identified and documented before deciding how to handle them during preprocessing.



## Step 2 – Demonstrate the Concept

### 1. Simple Example

Consider a dataset containing customer information.

| Customer | Age | City |
|---|---:|---|
| A | 25 | Chennai |
| B | NaN | Salem |
| C | 30 | NaN |

Here, the Age value for Customer B and the City value for Customer C are missing.

### 2. Real-World Example

In an online retail dataset, some transaction records may not contain a Customer ID or product Description.

These missing values may limit certain types of analysis.

### 3. Business Example

A retailer wants to analyze purchasing activity by customer. If Customer ID is missing for some records, those records may not be usable for customer-level grouping, even though they may still provide useful transaction information.

### 4. AI/ML Use Case

Missing values can cause errors in some machine learning algorithms or reduce the number of usable records.

Depending on the feature and project objective, preprocessing may involve imputation, excluding selected records, or using methods that support missing values.

### 5. Why This Concept Is Important in Data Analysis

Identifying missing values helps assess data completeness and plan suitable preprocessing. Missing values should not be filled or removed without understanding their extent and possible meaning.


```python

# Identify missing values in the dataset

# Count missing values in each feature
missing_counts = df.isnull().sum()

# Calculate missing-value percentages
missing_percentages = (
    df.isnull().sum() / len(df) * 100
)

# Create a summary table
missing_summary = pd.DataFrame({
    "Feature": df.columns,
    "Missing_Count": missing_counts.values,
    "Missing_Percentage": missing_percentages.values
})

# Display features that contain missing values
missing_summary = missing_summary[
    missing_summary["Missing_Count"] > 0
].sort_values(
    "Missing_Count",
    ascending=False
)

print("Features Containing Missing Values:")
display(missing_summary.round(2))

# Calculate overall missing-value count
total_missing = int(df.isnull().sum().sum())

print("\nTotal Missing Values:", total_missing)
print(
    "Total Missing Percentage:",
    round(total_missing / df.size * 100, 2),
    "%"
)

```


## Step 4 – Code Explanation and Interpretation

### 1. What the Code Does

The code counts missing values in each feature, calculates their percentages, displays a summary of affected features, and calculates the overall missing-value count and percentage.

### 2. Why These Methods Were Used

- `df.isnull()` identifies missing entries.
- `sum()` counts missing values across rows or columns.
- Dividing the missing count by the total number of rows calculates the percentage for each feature.
- `df.size` returns the total number of cells in the dataset.
- Filtering with `Missing_Count > 0` displays only features containing missing values.

These methods help measure the extent of missing data.

### 3. What the Output Means

The summary shows which features contain missing values and how many entries are missing in each feature.

The overall missing percentage represents the proportion of missing cells across the entire dataset.

If the summary table is empty, no missing values were detected by this check.

### 4. Insights That Can Be Obtained

The results help identify features with incomplete information and determine which areas require further investigation.

The importance of missing values depends on the feature's purpose and the analysis being performed. Missing customer information, for example, may affect customer-level analysis differently from product descriptions.

### 5. Impact on the ML Pipeline

Missing values may affect feature selection, data preparation, and model training. The findings help plan an appropriate missing-value treatment strategy during preprocessing.

No missing values are filled, and no records are removed in this step.



# Topic 6: Identify Duplicates
## Step 1 – Understand the Concept

Duplicate records are rows that repeat information already present in a dataset.

Identifying duplicates involves:
- Counting exact duplicate rows.
- Inspecting duplicated records.
- Understanding how duplicate records may affect analysis.
- Determining whether repeated records are genuine duplicates or valid transactions.

Duplicate detection is an important data quality check because unintended duplicates can affect statistical summaries and business results.

Identifying duplicates does not automatically mean they should be removed.



## Step 2 – Demonstrate the Concept

### 1. Simple Example

Consider the following dataset:

| Product | Quantity |
|---|---:|
| Pen | 5 |
| Book | 3 |
| Pen | 5 |

The first and third rows contain identical values. They may be duplicate records, but their meaning should be checked before removal.

### 2. Real-World Example

An online retail dataset may contain repeated transaction rows because of repeated data entry, data imports, or legitimate repeated purchases.

Exact duplicate detection helps identify rows that need investigation.

### 3. Business Example

If an unintended duplicate transaction is counted twice, the business may overestimate transaction quantities or recorded sales value.

However, two identical-looking purchases can sometimes represent separate legitimate transactions.

### 4. AI/ML Use Case

Unintended duplicate records can cause certain observations to receive excessive weight during model training and may distort evaluation if related records appear in both training and testing datasets.

Duplicate checks can help improve data quality before model development.

### 5. Why This Concept Is Important in Data Analysis

Identifying duplicates helps assess dataset reliability and avoid potentially misleading statistical summaries or business conclusions.

Duplicates should be investigated using transaction identifiers and business context before deciding how to handle them.


```python

# Identify exact duplicate records

# Count all duplicate rows after their first occurrence
duplicate_count = df.duplicated().sum()

print("Total Exact Duplicate Rows:", duplicate_count)

# Display duplicate records for inspection
duplicate_records = df[df.duplicated(keep=False)]

print("\nDuplicate Records Including Their First Occurrences:")
display(duplicate_records.head(10))

# Calculate the percentage of duplicate rows
duplicate_percentage = (
    duplicate_count / len(df) * 100
    if len(df) > 0
    else 0
)

print(
    "\nDuplicate Row Percentage:",
    round(duplicate_percentage, 2),
    "%"
)

# Check repeated invoice identifiers separately
repeated_invoice_rows = df[
    df.duplicated(subset=["Invoice"], keep=False)
]

print(
    "\nRows With Repeated Invoice Identifiers:",
    len(repeated_invoice_rows)
)

print(
    "Note: Repeated invoice identifiers do not necessarily "
    "represent duplicate records."
)
```


## Step 4 – Code Explanation and Interpretation

### 1. What the Code Does

The code counts exact duplicate rows, displays sample duplicate records, calculates the duplicate-row percentage, and checks how many rows share an Invoice identifier with another row.

### 2. Why These Methods Were Used

- `df.duplicated()` identifies rows that repeat earlier rows across all columns.
- `sum()` counts duplicate rows.
- `df[df.duplicated(keep=False)]` displays all occurrences of duplicated rows, including the first occurrence.
- The percentage calculation measures the proportion of duplicate rows.
- `duplicated(subset=["Invoice"])` checks repeated invoice identifiers rather than exact row duplicates.

### 3. What the Output Means

The exact duplicate count indicates how many rows repeat an earlier row across all columns. The displayed records allow further inspection.

The repeated-invoice count is different: one invoice may legitimately contain multiple product lines. Therefore, repeated invoice identifiers alone do not establish that records are duplicates.

### 4. Insights That Can Be Obtained

The output helps assess whether exact duplicates exist and whether invoice identifiers repeat across multiple rows.

Actual findings should be documented from the executed results. Any suspected duplicate should be investigated before a decision is made.

### 5. Impact on the ML Pipeline

Unintended duplicates may affect feature distributions, model training, and evaluation. Detecting them supports reliable preprocessing and helps reduce the risk of biased results.

No rows are deleted or modified in this step.

# Step 1 – Understand the Concept: Detect Outliers

An outlier is a data value that is unusually far from other values in a dataset.

In the Online Retail dataset, the `Quantity` and `Price` columns can contain values that need further investigation.

Common methods for detecting outliers include:

- IQR (Interquartile Range): Uses the first quartile (Q1) and third quartile (Q3) to identify values outside a calculated range.
- Z-Score: Measures how far a value is from the mean in terms of standard deviations.
- Box Plot: Visually shows the distribution and potential outliers in numerical data.

For this topic, the IQR method will be used to identify potential outliers in `Quantity` and `Price`.

An outlier is not necessarily an error. It may represent a genuine bulk purchase, an expensive product, or an unusual transaction.

The original dataset will not be modified, and no outliers will be removed.

# Step 2 – Demonstrate the Concept: Detect Outliers

### Simple Example

Consider these values:

`10, 12, 13, 14, 15, 16, 100`

Most values are between 10 and 16, while 100 is much larger than the others. The value 100 may be a potential outlier and should be investigated.

### Real-World Example

An online store may have ordinary orders containing a few items and some orders containing hundreds of items. The larger quantities may be genuine bulk purchases rather than errors.

### Business Example

A retailer can investigate unusually large quantities or prices to identify bulk orders, pricing issues, or unusual transactions.

### AI/ML Use Case

Outlier detection can help identify unusual transactions and improve data-quality checks before training machine learning models. Some models are more sensitive to extreme values than others.

### Why This Is Important During Data Analysis

Outliers can affect statistical summaries, visualizations, and certain machine learning algorithms. Detecting them helps analysts decide whether further investigation or preprocessing is necessary.

The IQR method identifies potential outliers; it does not automatically determine whether they are incorrect.

```python
# Step 3 – Implement the Concept: Detect Outliers

# Select numerical features for outlier detection
numerical_features = ["Quantity", "Price"]

# Create a dictionary to store outlier boundaries
outlier_results = {}

for column in numerical_features:
    # Calculate the first and third quartiles
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    # Calculate the interquartile range
    IQR = Q3 - Q1

    # Calculate lower and upper boundaries
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify potential outliers
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    # Store the results
    outlier_results[column] = {
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Potential Outlier Count": len(outliers),
        "Potential Outlier Percentage": (
            len(outliers) / df[column].notna().sum() * 100
            if df[column].notna().sum() > 0 else 0
        )
    }

# Display the outlier analysis results
outlier_summary = pd.DataFrame(outlier_results).T

print("IQR Outlier Detection Summary:")
display(outlier_summary)
```

# Step 4 – Code Explanation and Interpretation

### 1. What the Code Does

The code applies the Interquartile Range (IQR) method to the `Quantity` and `Price` columns.

It calculates the quartiles, IQR, lower and upper boundaries, and the number and percentage of potential outliers.

### 2. Why These Methods Were Used

- `quantile(0.25)` calculates the first quartile (Q1).
- `quantile(0.75)` calculates the third quartile (Q3).
- `IQR = Q3 - Q1` measures the spread of the middle 50% of values.
- The `1.5 × IQR` rule establishes the lower and upper boundaries.
- Boolean conditions identify values outside these boundaries.
- `len(outliers)` counts the potential outlier rows.
- `notna().sum()` counts non-missing values, providing the denominator for the outlier percentage.
- `DataFrame()` organizes the results into a readable summary table.

### 3. What the Output Means

- Q1 and Q3 describe the lower and upper quartiles.
- IQR measures the spread between those quartiles.
- The lower and upper boundaries define the IQR-based outlier range.
- Potential Outlier Count shows the number of rows outside the boundaries.
- Potential Outlier Percentage shows the proportion of non-missing values flagged as potential outliers.

There are three common types of correlation:

- Positive Correlation: Both variables tend to increase or decrease together.

```python
# Step 3 – Implement the Concept: Recommendations for the Next Preprocessing Stage

# Collect data-quality information for preprocessing planning
recommendations = []

# 1. Missing-value assessment
missing_counts = df.isnull().sum()
missing_columns = missing_counts[missing_counts > 0]

if not missing_columns.empty:
    recommendations.append({
        "Area": "Missing Values",
        "Finding": f"{len(missing_columns)} columns contain missing values.",
        "Recommendation": "Investigate missingness and decide how to handle each column."
    })
else:
    recommendations.append({
        "Area": "Missing Values",
        "Finding": "No missing values detected.",
        "Recommendation": "Confirm data completeness for the intended task."
    })

# 2. Duplicate-row assessment
duplicate_count = df.duplicated().sum()

recommendations.append({
    "Area": "Duplicate Rows",
    "Finding": f"{duplicate_count} duplicate rows detected.",
    "Recommendation": "Verify repeated records before deciding whether removal is appropriate."
})

# 3. Data type assessment
recommendations.append({
    "Area": "Data Types",
    "Finding": f"InvoiceDate dtype: {df['InvoiceDate'].dtype}.",
    "Recommendation": "Validate date and numerical column types before further analysis."
})

# 4. Outlier assessment
for column in ["Quantity", "Price"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    recommendations.append({
        "Area": f"{column} Outliers",
        "Finding": f"{outlier_count} potential outlier rows detected.",
        "Recommendation": "Investigate flagged values before deciding how to handle them."
    })

# 5. Negative quantity assessment
negative_quantity_count = (df["Quantity"] < 0).sum()

recommendations.append({
    "Area": "Negative Quantities",
    "Finding": f"{negative_quantity_count} rows contain negative quantities.",
    "Recommendation": "Check transaction context, including possible returns or adjustments."
})

# Display the recommendations
recommendations_df = pd.DataFrame(recommendations)

print("Next-Stage Preprocessing Recommendations:")
display(recommendations_df)
```
