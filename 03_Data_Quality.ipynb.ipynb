```python
import pandas as pd
import numpy as np

file_path = "online_retail_II.xlsx"
df = pd.read_excel(file_path, sheet_name="Year 2009-2010")

print("Dataset loaded successfully!")
display(df.head())
```

# Missing Values

## Topic 1: Number of Missing Values

### Step 1: Understand the Concept

Missing values are values that are not available or recorded in a dataset. In Python, they are commonly represented as `NaN`, `None`, or `NaT`, depending on the data type.

Identifying missing values is the first step in assessing data quality. Missing information can affect calculations, visualizations, statistical analysis, and machine learning models.

**Why do we check missing values?**

- To understand how much information is missing from the dataset.
- To identify incomplete records.
- To decide whether missing values should be removed, filled, or investigated.
- To improve the reliability of data analysis and machine learning results.

**Important:** Missing values should not be removed automatically. The correct approach depends on the column, the reason for missing data, and the business problem.

## Step 2: Demonstrate the Concept

**Simple Example**

Consider a dataset containing customer information:

| Customer | Age | City |
|---|---:|---|
| A | 25 | Chennai |
| B | NaN | Madurai |
| C | 30 | NaN |
| D | NaN | Coimbatore |

There are three missing values in this example: two in the Age column and one in the City column.

**Real-World Example**

In an online shopping dataset, a transaction may have a missing product description or customer ID.

**Business Example**

If customer information is missing, a business may find it difficult to understand customer purchasing behaviour.

**AI/ML Use Case**

Missing values can cause errors in some machine learning algorithms or reduce model quality if they are not handled appropriately.

**Importance in Data Analysis**

Counting missing values helps us understand how complete the dataset is before moving to deeper analysis.

```python
# Count the total number of missing values in the dataset
total_missing_values = df.isnull().sum().sum()

print("Total Number of Missing Values:", total_missing_values)
```

## Code Explanation

**1. `df.isnull()`**

Checks every cell in the DataFrame and identifies missing values.

- `True` means the value is missing.
- `False` means the value is present.

**2. `df.isnull().sum()`**

Counts missing values separately for each column because `True` is treated as `1` and `False` as `0` during summation.

**3. `.sum().sum()`**

- The first `.sum()` counts missing values in each column.
- The second `.sum()` adds those column counts together to calculate the total number of missing values in the entire dataset.

**4. `print()`**

Displays the total missing-value count.

**Expected Output**

The output will look like this:

`Total Number of Missing Values: ...`

The actual number depends on the dataset loaded. The ellipsis is a placeholder, not a value to type into your code.

**What does the result mean?**

- If the result is `0`, pandas has not detected any missing values represented as standard null values.
- If the result is greater than `0`, the dataset contains missing values that need further investigation.

**Impact on the ML Pipeline**

This check helps identify whether missing-data handling may be needed during data preprocessing. It does not modify or delete any data.

# Topic 2: Percentage of Missing Values

## Step 1: Understand the Concept

The percentage of missing values tells us what proportion of the dataset's values are missing.

It is calculated by dividing the number of missing values by the total number of values and multiplying the result by 100.

**Formula:**

Missing Value Percentage = (Number of Missing Values / Total Number of Values) × 100

**Why do we calculate it?**

- To measure data completeness.
- To compare missing data across datasets or columns.
- To identify columns with a high percentage of missing values.
- To help decide how missing values should be handled.

A high missing percentage does not automatically mean that a column should be deleted. The decision depends on the column's importance and the business objective.

## Step 2: Demonstrate the Concept

**Simple Example**

Suppose a column contains 100 customer records, and 20 values are missing.

Missing Value Percentage = (20 / 100) × 100 = 20%

This means that 20% of the values in that column are missing.

**Real-World Example**

An online retail dataset may contain 1,000 customer IDs, of which 150 are missing.

**Business Example**

If many customer IDs are missing, the business may have difficulty analysing individual customer purchasing behaviour.

**AI/ML Use Case**

A high percentage of missing values in an important feature may affect model training and may require a suitable preprocessing strategy.

**Importance in Data Analysis**

Percentages help compare columns fairly, especially when interpreting missing data across different fields.

```python
# Calculate the total number of cells in the dataset
total_values = df.size

# Calculate the total number of missing values
total_missing_values = df.isnull().sum().sum()

# Calculate the percentage of missing values
if total_values > 0:
    missing_percentage = (total_missing_values / total_values) * 100
else:
    missing_percentage = 0

print(f"Total Values: {total_values}")
print(f"Missing Values: {total_missing_values}")
print(f"Missing Value Percentage: {missing_percentage:.2f}%")
```

## Code Explanation

**1. `df.size`**

Returns the total number of cells in the DataFrame, including rows multiplied by columns.

For example, a DataFrame with 100 rows and 5 columns contains 500 cells.

**2. `df.isnull().sum().sum()`**

Counts all missing values across the entire dataset.

**3. `if total_values > 0:`**

Checks that the dataset contains at least one cell before performing the division. This avoids division by zero.

**4. `(total_missing_values / total_values) * 100`**

Calculates the percentage of cells that contain missing values.

**5. `else: missing_percentage = 0`**

If the dataset has no cells, the code assigns zero to the result rather than attempting a division by zero. In that situation, the percentage is only a fallback value and should not be interpreted as evidence of complete data.

**6. `:.2f`**

Formats the percentage to two decimal places.

**7. `print()`**

Displays the total number of cells, the missing-value count, and the calculated percentage.

**What does the output mean?**

- A low percentage indicates that relatively few cells are missing.
- A high percentage indicates that more of the dataset is incomplete.
- A percentage of `0.00%` means no standard missing values were detected in the cells checked.

**Important Note**

This calculation measures missing cells across the entire dataset. In the next topic, we will calculate missing values separately for each column to identify which fields need attention.

**Impact on the ML Pipeline**

This measurement supports data-cleaning decisions before feature preparation and model training. It does not change the original dataset.

Missing values by column means counting the number of missing values separately for every column in a dataset.

Each column may contain a different number of missing values. Some columns may have no missing values, while others may have many.

**Why do we check missing values by column?**

- To identify columns with incomplete information.
- To find which columns require further investigation.
- To compare data completeness across different features.
- To decide whether missing values should be retained, filled, or removed.

**Important:** A column with many missing values is not automatically useless. Its business importance and the reason for the missing data should also be considered.

## Step 2: Demonstrate the Concept

**Simple Example**

Consider the following customer dataset:

| Name | Age | City |
|---|---:|---|
| Arun | 25 | Chennai |

**4. `.sort_values(ascending=False)`**

Sorts the counts from highest to lowest so columns with the most missing values appear first.

- `ascending=False` means descending order.
- Columns with zero missing values will appear toward the bottom.

**5. `to_frame(name="Missing Count")`**

Converts the Series into a DataFrame and names the resulting column `Missing Count`, making the output easier to read.

**6. `display()`**

Shows the result as a formatted table in Jupyter Notebook.

**What does the output mean?**

Each row shows a column from the retail dataset and its number of missing values. Check which columns have the highest counts and which have zero.

**Important Note**

This code counts missing values but does not change or remove them. It detects standard null values; empty strings and other unusual placeholders may need separate checks.

**Impact on the ML Pipeline**

Column-wise missing-value analysis helps guide preprocessing decisions, including whether a feature needs imputation or further investigation. These decisions should be made based on the data and the intended prediction task.

# Topic 4: Missing Value Patterns

## Step 1: Understand the Concept

Missing value patterns describe how missing values are distributed across rows and columns in a dataset.

They help us understand whether missing values occur independently, appear together in multiple columns, or are concentrated in particular records.

**Common Patterns**

- **Single-column missingness:** Missing values occur mainly in one column.
- **Multiple-column missingness:** Several columns contain missing values in the same rows.
- **Row-level missingness:** Some records contain many missing values.
- **Repeated missingness:** The same combination of columns is missing across multiple records.

**Why do we investigate missing value patterns?**

- To identify repeated data-quality issues.
- To understand which records are incomplete.
- To investigate possible reasons for missing information.
- To choose an appropriate missing-data handling strategy.

**Important:** A missing-value pattern alone does not prove why data is missing or whether the missingness is random. Additional context may be required.

## Step 2: Demonstrate the Concept

**Simple Example**

Consider this customer dataset:

| Customer | Age | City | Income |
|---|---:|---|---:|
| A | 25 | Chennai | 40000 |
| B | NaN | Madurai | 35000 |
| C | NaN | NaN | NaN |
| D | 30 | Chennai | NaN |

In this example:

- Customer B has a missing Age.
- Customer C has missing values in three columns.
- Customer D has a missing Income.

These records show different missing-value patterns.

**Real-World Example**

In online retail data, some transactions may have a missing Customer ID while product and transaction details are available.

**Business Example**

If many records have missing customer information, the business may need to investigate how customer details are captured during transactions.

**AI/ML Use Case**

Understanding missingness patterns helps data scientists select suitable preprocessing methods and avoid blindly deleting useful records.

**Importance in Data Analysis**

Pattern analysis provides more detail than a simple missing-value count and helps identify records that may need closer investigation.

```python
# Create a Boolean mask showing missing values
missing_mask = df.isnull()

# Count missing values in each row
missing_per_row = missing_mask.sum(axis=1)

# Count rows by the number of missing values they contain
print("Number of Rows by Missing-Value Count:")
display(missing_per_row.value_counts().sort_index().to_frame(
    name="Number of Rows"
))

# Show the most common missing-value combinations
print("\nMost Common Missing-Value Patterns:")
display(
    missing_mask.astype(int)
    .value_counts()
    .head(10)
    .to_frame(name="Number of Rows")
)

# Show rows that contain at least one missing value
print("\nSample Rows Containing Missing Values:")
display(df.loc[missing_per_row > 0].head(10))
```

## Code Explanation

**1. `df.isnull()`**

Creates a Boolean DataFrame in which `True` represents a missing value and `False` represents a present value.

**2. `missing_mask.sum(axis=1)`**

Counts missing values in each row.

- `axis=1` means the calculation is performed across columns for every row.
- A result of `0` means that row has no standard missing values.
- A result of `3` means that row has three missing values.

**3. `missing_per_row.value_counts()`**

Counts how many rows contain each possible number of missing values.

For example, it may show how many rows have zero, one, two, or more missing values.

**4. `.sort_index()`**

Sorts the row-level missing-value counts in ascending order of the number of missing values.

**5. `missing_mask.astype(int)`**

Converts Boolean values into integers:

- `True` becomes `1`.
- `False` becomes `0`.

This makes each row's missing-value combination easier to count.

**6. `.value_counts().head(10)`**

Counts repeated combinations of missing and present values, then displays the ten most frequent patterns.

Each combination represents the missingness pattern across the dataset's columns. The displayed combination contains `1` for a missing value and `0` for a present value.

**7. `.to_frame(name="Number of Rows")`**

Converts the counts into a DataFrame with a clear column name.

**8. `df.loc[missing_per_row > 0].head(10)`**

Selects rows with at least one missing value and displays up to ten examples for inspection.

**What does the output mean?**

- The first table shows how many rows contain different numbers of missing values.
- The second table shows the most frequent combinations of missing values across columns.
- The third table displays example records that need further investigation.

**Performance Note**

For a large dataset with many columns, counting every distinct missing-value combination can be expensive. Here, `.head(10)` limits the displayed results, but pandas still needs to calculate the combinations before selecting the ten most common.

**Impact on the ML Pipeline**

Missing-value patterns help guide data-cleaning and preprocessing decisions. Investigating the reason for missingness is important before choosing imputation or record-removal methods.

# Duplicate Records

## Topic 5: Identify Duplicates

### Step 1: Understand the Concept

Duplicate records are records that repeat data already present in a dataset.

Identifying duplicates is an important part of data quality assessment because repeated records may affect transaction counts, revenue calculations, and statistical analysis.

**Types of Duplicates**

- **Completely duplicated rows:** Every column has the same value as another row.
- **Potential duplicate records:** Selected columns match, but other columns may differ.

**Why do we identify duplicates?**

- To detect repeated records.
- To investigate possible data-entry or data-import issues.
- To prevent incorrect counts and calculations.
- To decide whether records need further investigation.

**Important:** A repeated record is not always an error. Before removing duplicates, we must understand the dataset and confirm whether the repeated information represents a genuine transaction or an accidental repetition.

## Step 2: Demonstrate the Concept

**Simple Example**

| Invoice | Product | Quantity |
|---|---|---:|
| 1001 | Pen | 2 |
| 1002 | Book | 1 |
| 1001 | Pen | 2 |

The first and third rows contain identical values in all three displayed columns. They would be duplicates if these were the complete columns being checked.

**Real-World Example**

A retail system may import the same transaction record twice because of an import or processing issue.

**Business Example**

Repeated transaction rows may cause a business to overestimate the number of items sold or the value of sales.

**AI/ML Use Case**

Duplicates can introduce unintended weighting into a dataset and influence model training or evaluation.

**Importance in Data Analysis**

Duplicate identification helps us investigate data reliability before performing detailed analysis.

```python
# Identify rows that are duplicates of earlier rows
duplicate_mask = df.duplicated()

# Display whether each row is a duplicate
print("Duplicate Status for the First 10 Rows:")
display(duplicate_mask.head(10).to_frame(name="Is Duplicate"))

# Display sample duplicate records
print("\nSample Duplicate Records:")
display(df.loc[duplicate_mask].head(10))
```

## Code Explanation

**1. `df.duplicated()`**

Checks whether each row duplicates a row that appeared earlier in the DataFrame.

By default, pandas compares all columns and marks only subsequent occurrences as duplicates.

- `True` means the row has appeared before with the same values in all compared columns.
- `False` means pandas has not found an earlier identical row.

**2. `duplicate_mask`**

Stores the Boolean results so we can reuse them for filtering and analysis.

**3. `duplicate_mask.head(10)`**

Displays the duplicate status for the first ten rows. This is a small preview and does not guarantee that those rows contain duplicates.

**4. `.to_frame(name="Is Duplicate")`**

Converts the Boolean Series into a one-column DataFrame with a clear column name.

**5. `df.loc[duplicate_mask]`**

Selects the rows marked as duplicates.

**6. `.head(10)`**

Displays up to ten duplicate records for inspection.

**What does the output mean?**

If the sample duplicate table is empty, pandas did not find any duplicate rows in the selected result. This does not mean there are no potential duplicates based on only some columns.

**Important Note**

This code identifies completely duplicated rows across all columns. It does not delete or modify any records. We will count the duplicates separately in the next topic.

**Impact on the ML Pipeline**

Identifying duplicate records helps prevent unintended repetition from distorting data summaries or model training. Any removal should happen only after investigating whether the repeated records are truly redundant.

# Topic 6: Count Duplicates

## Step 1: Understand the Concept

Counting duplicates means calculating the total number of rows that repeat an earlier row across all columns.

In pandas, `duplicated()` identifies duplicate occurrences, and `sum()` counts how many rows are marked as duplicates.

**Why do we count duplicates?**

- To measure the number of repeated records.
- To understand the extent of possible duplication in a dataset.
- To support data-cleaning decisions.
- To investigate whether repeated rows may affect business calculations.

**Important:** The duplicate count does not include the first occurrence of a repeated row. It counts only subsequent occurrences by default.

## Step 2: Demonstrate the Concept

**Simple Example**

| Invoice | Product | Quantity |
|---|---|---:|
| 1001 | Pen | 2 |
| 1002 | Book | 1 |
| 1001 | Pen | 2 |
| 1001 | Pen | 2 |

The first occurrence of invoice 1001 is not counted as a duplicate. The second and third occurrences are duplicates.

Therefore, the duplicate count is 2.

**Real-World Example**

An online shopping system may accidentally import a transaction multiple times.

**Business Example**

Duplicate transactions may inflate transaction counts and sales totals if they are included without investigation.

**AI/ML Use Case**

Counting duplicates helps determine whether repeated observations require investigation before training a machine learning model.

**Importance in Data Analysis**

The count gives us a quick measure of repeated records and helps us decide whether further investigation is necessary.

```python
# Count completely duplicated rows
duplicate_count = df.duplicated().sum()

# Calculate the total number of rows
total_rows = len(df)

# Calculate the percentage of duplicate rows
if total_rows > 0:
    duplicate_percentage = (duplicate_count / total_rows) * 100
else:
    duplicate_percentage = 0

print("Total Rows:", total_rows)
print("Duplicate Rows:", duplicate_count)
print(f"Duplicate Row Percentage: {duplicate_percentage:.2f}%")
```

## Code Explanation

**1. `df.duplicated()`**

Checks every row against earlier rows using all columns. Subsequent occurrences of identical rows are marked as `True`.

**2. `.sum()`**

Counts the `True` values to calculate the total number of duplicate rows.

**3. `len(df)`**

Returns the total number of rows in the dataset.

**4. `if total_rows > 0:`**

Checks whether the dataset contains rows before calculating the percentage.

**5. `(duplicate_count / total_rows) * 100`**

Calculates the percentage of rows marked as duplicates relative to the total number of rows.

**6. `else: duplicate_percentage = 0`**

Provides a fallback when the DataFrame has no rows. If the dataset is empty, this zero is not a meaningful measurement of duplicate prevalence.

**7. `print()` and `:.2f`**

Display the results. The formatting `:.2f` shows the percentage to two decimal places.

**What does the output mean?**

- `Total Rows` shows the dataset's row count.
- `Duplicate Rows` shows subsequent rows identical across all columns.
- `Duplicate Row Percentage` shows the proportion of rows marked as duplicates.

A higher percentage indicates more repeated rows, but it does not automatically prove that the records are incorrect.

**Impact on the ML Pipeline**

This metric helps assess whether duplicate observations could influence data analysis or model training. Investigate the business meaning of the rows before deciding whether to remove them.

# Topic 8: Identify Potential Duplicate Records

## Step 1: Understand the Concept

Potential duplicate records are records that match across selected columns but may differ in other columns.

Unlike completely duplicated rows, potential duplicates are identified using a subset of columns that are meaningful for a particular business problem.

**Why do we identify potential duplicates?**

- To detect records that may represent the same transaction or event.
- To investigate repeated invoice and product combinations.
- To identify possible data-entry or data-processing issues.
- To improve the reliability of transaction-level analysis.

**Important:** There is no universal rule for identifying potential duplicates. The columns used for comparison must be selected according to the dataset structure and business requirements. A matching invoice and product code may be legitimate, so the results must be investigated before changing the data.

## Step 2: Demonstrate the Concept

**Simple Example**

| Invoice | StockCode | Quantity | Price |
|---|---|---:|---:|
| 1001 | P101 | 2 | 10 |
| 1001 | P101 | 3 | 10 |
| 1002 | P102 | 1 | 50 |

The first two records have the same Invoice and StockCode, but their quantities differ.

They are potential duplicates based on the selected columns, but they are not completely duplicated rows.

**Real-World Example**

An online retailer may have multiple records for the same product within one invoice.

**Business Example**

Investigating repeated invoice-product combinations can help identify data-entry issues or understand how transaction lines are recorded.

**AI/ML Use Case**

Potential duplicate checks help analysts investigate repeated observations before creating features or training a model.

**Importance in Data Analysis**

Checking selected columns can reveal suspicious patterns that a full-row duplicate check will not detect.

```python
# Select columns for potential duplicate checking
comparison_columns = ["Invoice", "StockCode"]

# Check that the required columns exist
available_columns = [
    column for column in comparison_columns
    if column in df.columns
]

if len(available_columns) == len(comparison_columns):
    # Identify repeated Invoice and StockCode combinations
    potential_duplicate_mask = df.duplicated(
        subset=comparison_columns,
        keep=False
    )

    # Extract all records belonging to repeated combinations
    potential_duplicates = df.loc[potential_duplicate_mask].copy()

    print("Potential Duplicate Records:",
          len(potential_duplicates))

    display(
        potential_duplicates
        .sort_values(comparison_columns)
        .head(20)
    )
else:
    print("Required comparison columns are not available.")
```

## Code Explanation

**1. `comparison_columns = ["Invoice", "StockCode"]`**

Defines the columns used to identify potential duplicates. Records matching on both columns are considered part of a repeated combination.

**2. `available_columns`**

Checks which selected columns actually exist in the DataFrame. This helps prevent an error if a required column is missing.

**3. `if len(available_columns) == len(comparison_columns):`**

Continues the analysis only if all requested columns are available.

**4. `df.duplicated(subset=comparison_columns, keep=False)`**

Checks for repeated combinations using only Invoice and StockCode.

- `subset=comparison_columns` restricts the comparison to the selected columns.
- `keep=False` marks every row belonging to a repeated combination as `True`, including the first occurrence.

**5. `df.loc[potential_duplicate_mask].copy()`**

Extracts all records flagged as potential duplicates into a separate DataFrame.

**6. `len(potential_duplicates)`**

Counts all records belonging to repeated combinations. This is a record count, not the number of distinct duplicate groups.

**7. `.sort_values(comparison_columns)`**

Groups similar Invoice and StockCode combinations together in the displayed results, making them easier to inspect.

**8. `.head(20)`**

Displays up to 20 records for review.

**What does the output mean?**

The table contains records where the same invoice and stock code combination appears more than once. Other fields, such as Quantity or Price, may be different.

These records require investigation; they should not automatically be treated as errors or deleted.

**Impact on the ML Pipeline**

Potential duplicate checks help analysts assess repeated transaction patterns before aggregation, feature engineering, and model training. The correct comparison columns depend on the intended analysis.

# Invalid Values

## Topic 9: Negative Values Where Inappropriate

### Step 1: Understand the Concept

Negative values are numerical values that are less than zero. In data quality assessment, we check whether negative values are valid for a particular column or represent possible data errors.

**Examples in an Online Retail Dataset**

- **Quantity:** Negative quantities may represent returned or cancelled items.
- **Price:** Negative prices may require investigation because ordinary product selling prices are generally expected to be non-negative.
- **Sales Amount:** A negative amount may be valid when representing a refund or return, depending on the calculation method.

**Why do we check negative values?**

- To identify unusual numerical values.
- To distinguish valid business events from possible data errors.
- To prevent incorrect calculations.
- To support reliable data cleaning and analysis.

**Important:** Do not automatically replace negative values with zero or delete the records. First investigate the column's meaning and the business context.

## Step 2: Demonstrate the Concept

**Simple Example**

| Product | Quantity | Price |
|---|---:|---:|
| Pen | 5 | 10 |
| Book | -2 | 50 |
| Bag | 1 | -20 |

- The negative quantity for the book may represent a return.
- The negative price for the bag should be investigated.

**Real-World Example**

An online customer returns two items, so the transaction system may record a negative quantity to represent the return.

**Business Example**

If negative values are not interpreted correctly, sales reports may become misleading. However, removing valid returns can also make the sales figures inaccurate.

**AI/ML Use Case**

Negative-value checks help identify unusual data before feature engineering and model training.

**Simple Example**

| Product | Quantity | Price |
|---|---:|---:|

```python
# Define numerical columns to inspect
numeric_columns = ["Quantity", "Price"]

for column in numeric_columns:
    if column in df.columns:
        print(f"\nUnexpected Value Summary for {column}:")

        # Calculate descriptive statistics
        display(df[column].describe().to_frame(name="Statistics"))

        # Calculate the 1.5 IQR bounds
        q1 = df[column].quantile(0.25)
        q3 = df[column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        # Identify values outside the IQR bounds
        unusual_mask = (
            (df[column] < lower_bound)
            | (df[column] > upper_bound)
        )

        unusual_values = df.loc[unusual_mask]

        print("Lower Bound:", lower_bound)
        print("Upper Bound:", upper_bound)
        print("Number of Unusual Records:", len(unusual_values))

        display(unusual_values.head(10))
```

## Code Explanation

**1. `numeric_columns = ["Quantity", "Price"]`**

Selects the numerical columns to inspect for unusual values.

Calculate the first quartile (Q1) and third quartile (Q3).

- Q1 is the 25th percentile.
- Q3 is the 75th percentile.

**5. `iqr = q3 - q1`**

Calculates the interquartile range (IQR), which measures the spread of the middle 50% of the data.

**6. Calculate the lower and upper bounds**

The code uses the common 1.5 × IQR rule:

- Lower bound = Q1 − 1.5 × IQR
- Upper bound = Q3 + 1.5 × IQR

**7. `unusual_mask`**

Identifies values below the lower bound or above the upper bound.

**8. `df.loc[unusual_mask]`**

Selects the records flagged by the IQR rule.

**9. `len(unusual_values)` and `head(10)`**

Counts the flagged records and displays up to ten examples.

# Topic 15: Empty Strings

## Step 1: Understand the Concept

An empty string is a text value that contains no characters. A whitespace-only string contains spaces but no meaningful text.

Examples include:

- `""` — an empty string.
- `"   "` — a string containing only spaces.
- `"Chennai"` — a non-empty string.

Empty strings and whitespace-only strings are different from standard missing values such as `NaN` or `None`.

**Why do we check empty strings?**

- To identify incomplete text fields.
- To distinguish blank text from standard missing values.
- To improve the quality of categorical data.
- To prevent blank values from being treated as meaningful categories during analysis or machine learning.

**Important:** An empty string may represent missing information, but the correct treatment depends on the meaning of the column.

## Step 2: Demonstrate the Concept

**Simple Example**

| Customer | City |
|---|---|
| Arun | Chennai |
| Priya | `""` |
| Ravi | `"   "` |
| Meena | Madurai |

**Real-World Example**

**AI/ML Use Case**

Empty strings may be treated as a separate category during encoding, even when they actually represent missing information.

**Importance in Data Analysis**

Checking empty strings helps detect incomplete text fields that a standard missing-value check may not identify.

```python
# Identify text-like columns
text_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns

empty_string_counts = {}

for column in text_columns:
    text_values = df[column].astype("string")

    # Detect empty strings and whitespace-only strings
    empty_mask = (
        text_values.notna()
        & text_values.str.strip().eq("")
    )

    empty_string_counts[column] = empty_mask.sum()

# Display counts for each text-like column
empty_string_summary = pd.Series(
    empty_string_counts,
    name="Empty or Whitespace-Only Count"
)

print("Empty Strings by Text Column:")
display(empty_string_summary.to_frame())

print(
    "\nTotal Empty or Whitespace-Only Values:",
    int(empty_string_summary.sum())
)
```

**2. `.columns`**

Returns the names of the selected columns.

**3. `empty_string_counts = {}`**

Creates an empty dictionary to store the number of empty or whitespace-only values in each text column.

**4. `df[column].astype("string")`**

Creates a string-typed version of the column for inspection. The original DataFrame is not changed.

Excludes standard missing values from the empty-string check.

- `str.strip()` removes whitespace from the beginning and end of each string.

**8. `pd.Series(empty_string_counts, name="Empty or Whitespace-Only Count")`**

Converts the dictionary into a pandas Series with a descriptive name.

**9. `.to_frame()` and `display()`**

Convert and display the summary in a notebook-friendly table.

**10. `empty_string_summary.sum()`**

**What does the output mean?**

A count of zero means no empty or whitespace-only strings were detected in the selected text-like columns.

**Important Note**

# Topic 16: Null-Like Values

## Step 1: Understand the Concept

Null-like values are text values that may be used to represent missing or unavailable information.

Common examples include:

- `"NULL"`
- `"None"`
- `"N/A"`
- `"NA"`
- `"unknown"`
- `"not available"`

These values are stored as ordinary text strings, so pandas may not recognize them as standard missing values such as `NaN` or `None`.

**Why do we check null-like values?**

- To identify missing information represented as text.
- To improve data consistency.
- To prevent incorrect categories during analysis.
- To prepare reliable data for machine learning.

**Important:** A null-like text value should only be treated as missing after checking the meaning of the column and the dataset's rules.

## Step 2: Demonstrate the Concept

**Simple Example**

| Customer | City |
|---|---|
| Arun | Chennai |
| Priya | N/A |
| Ravi | Unknown |
| Meena | Madurai |

The values `"N/A"` and `"Unknown"` may indicate that the city information is unavailable.

**Real-World Example**

A registration form may save `"N/A"` when a customer does not provide an optional field.

**Business Example**

If unavailable product information is stored as `"NULL"`, a report may incorrectly count it as a valid product category.

**AI/ML Use Case**

Null-like values can create misleading categories when categorical features are encoded for machine learning models.

**Importance in Data Analysis**

Identifying these values helps distinguish genuine information from text that may represent missing data.

```python
# Define common text placeholders that may represent missing values
null_like_values = [
    "NULL",
    "None",
    "N/A",
    "NA",
    "N.A.",
    "not available",
    "unknown"
]

# Select text-like columns
text_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns

null_like_summary = {}

for column in text_columns:
    text_values = df[column].astype("string")

    # Normalize text for comparison without changing the original data
    normalized_values = text_values.str.strip().str.casefold()

    # Identify values matching the specified placeholders
    null_like_mask = normalized_values.isin(
        [value.casefold() for value in null_like_values]
    )

    null_like_summary[column] = int(null_like_mask.sum())

# Display counts by column
summary = pd.Series(
    null_like_summary,
    name="Null-Like Value Count"
)

print("Null-Like Values by Text Column:")
display(summary.to_frame())

print(
    "\nTotal Null-Like Values:",
    int(summary.sum())
)
```

## Code Explanation

**1. `null_like_values`**

Creates a list of common text placeholders that may represent missing information.

**2. `df.select_dtypes(include=["object", "string", "category"])`**

Selects columns that may contain text values.

**3. `df[column].astype("string")`**

Converts each selected column to pandas' string data type for checking. The original DataFrame remains unchanged.

**4. `.str.strip()`**

Removes spaces from the beginning and end of text values. For example, `" NULL "` becomes `"NULL"` for comparison.

**5. `.str.casefold()`**

Makes text comparison case-insensitive. For example, `"NULL"`, `"null"`, and `"Null"` are treated as equivalent.

**6. `.isin()`**

Checks whether each value matches one of the specified null-like placeholders.

**7. `null_like_mask.sum()`**

Counts the matching values in each column.

**8. `pd.Series(null_like_summary, name="Null-Like Value Count")`**

Creates a summary containing the number of detected null-like values in each text column.

**9. `display(summary.to_frame())`**

Displays the summary as a table in Jupyter Notebook.

**10. `summary.sum()`**

Calculates the total number of matching cells across the selected columns.

**What does the output mean?**

- A positive count indicates that the column contains one or more specified placeholders.
- A count of zero indicates that none of the listed placeholders were detected in that column.
- The total shows the number of matching cells across all selected text columns.

**Important Note**

The result depends on the placeholders included in `null_like_values`. Other spellings or placeholders will not be detected unless they are added to the list. Some values, such as `"unknown"`, may be meaningful in certain datasets, so verify them before replacing anything.

**Impact on the ML Pipeline**

Checking null-like values helps standardize missing-data handling before feature engineering, encoding, and model training. This code only identifies values; it does not replace or delete them.
