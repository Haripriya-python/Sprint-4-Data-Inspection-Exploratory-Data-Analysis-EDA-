
## Question 2 – Why Is EDA Performed Before Machine Learning?

### Coding Exercise:

Using the `online_retail_II.xlsx` dataset:

1. Check the dataset's shape.
2. Identify missing values and duplicate records.
3. Display descriptive statistics for numerical features.
4. Identify the numerical columns that contain negative values.
5. Print a summary of the data quality issues identified before preparing the dataset for Machine Learning.


```python

import pandas as pd

# Load the dataset
df = pd.read_excel(
    "online_retail_II.xlsx",
    sheet_name="Year 2009-2010"
)

# 1. Check the dataset shape
print("Dataset Shape:", df.shape)

# 2. Identify missing values and duplicate records
print("\nMissing Values:")
print(df.isnull().sum())

print("\nNumber of Duplicate Records:", df.duplicated().sum())

# 3. Display descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())

# 4. Identify numerical columns containing negative values
print("\nNegative Values in Numerical Columns:")

numeric_df = df.select_dtypes(include="number")

for column in numeric_df.columns:
    negative_count = (numeric_df[column] < 0).sum()

    if negative_count > 0:
        print(f"{column}: {negative_count} negative values")

# 5. Summarize data quality issues
print("\nData Quality Summary:")
print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("Total Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Records:", df.duplicated().sum())
```


## Question 3 – Univariate, Bivariate, and Multivariate Analysis

### Coding Exercise:

Using the `online_retail_II.xlsx` dataset:

1. Perform univariate analysis on the `Quantity` column using descriptive statistics and a histogram.
2. Perform bivariate analysis on the `Quantity` and `Price` columns using a scatter plot.
3. Perform multivariate analysis using `Quantity`, `Price`, and `Country` to compare average quantities across countries.
4. Display the results and create appropriate visualizations.


```python

import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_excel(
    "online_retail_II.xlsx",
    sheet_name="Year 2009-2010"
)

# Prepare the required columns for analysis
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

# 1. Univariate Analysis: Quantity
print("Univariate Analysis - Quantity")
display(df["Quantity"].describe())

plt.figure(figsize=(8, 5))
plt.hist(df["Quantity"].dropna(), bins=30)
plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()

# 2. Bivariate Analysis: Quantity vs Price
bivariate_df = df[["Quantity", "Price"]].dropna()

plt.figure(figsize=(8, 5))
plt.scatter(
    bivariate_df["Quantity"],
    bivariate_df["Price"],
    alpha=0.5
)
plt.title("Quantity vs Price")
plt.xlabel("Quantity")
plt.ylabel("Price")
plt.show()

# 3. Multivariate Analysis: Quantity, Price, and Country
multivariate_df = df[
    ["Quantity", "Price", "Country"]
].dropna()

country_summary = (
    multivariate_df.groupby("Country")
    .agg(
        Average_Quantity=("Quantity", "mean"),
        Average_Price=("Price", "mean"),
        Transaction_Count=("Country", "size")
    )
    .sort_values("Average_Quantity", ascending=False)
)

print("Multivariate Analysis by Country:")
display(country_summary.head(10))

# Visualize average quantity for the top 10 countries
top_countries = country_summary.head(10)

top_countries["Average_Quantity"].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Average Quantity by Country - Top 10")
plt.xlabel("Country")
plt.ylabel("Average Quantity")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```


## Question 6 – What Is an Outlier? Detect Outliers Using IQR

### Coding Exercise:

Using the `online_retail_II.xlsx` dataset:

1. Select the `Quantity` and `Price` columns for analysis.
2. Calculate Q1, Q3, and IQR for each column.
3. Calculate the lower and upper bounds using the IQR method.
4. Count the outliers in each column.
5. Display the outlier records.

Do not remove any outliers from the dataset.


```python

import pandas as pd

# Load the dataset
df = pd.read_excel(
    "online_retail_II.xlsx",
    sheet_name="Year 2009-2010"
)

# Select numerical features
features = ["Quantity", "Price"]

for column in features:
    values = pd.to_numeric(df[column], errors="coerce").dropna()

    # Calculate Q1, Q3, and IQR
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    # Calculate IQR bounds
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    # Identify outliers
    outlier_mask = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    )

    outliers = df[outlier_mask]

    print(f"\nOutlier Analysis: {column}")
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))

    print("\nFirst Five Outlier Records:")
    display(outliers.head())

# Confirm that the original dataset remains unchanged
print("\nOriginal Dataset Shape:", df.shape)
```


## Question 8 – What Is Correlation? Create a Correlation Heatmap

### Coding Exercise:

Using the `online_retail_II.xlsx` dataset:

1. Select the numerical features `Quantity` and `Price`.
2. Calculate the Pearson correlation between these features.
3. Display the correlation matrix.
4. Create a correlation heatmap using Seaborn.
5. Print the correlation value and identify whether it is positive, negative, or close to zero.

Remember that correlation does not prove causation.


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_excel(
    "online_retail_II.xlsx",
    sheet_name="Year 2009-2010"
)

# Select numerical features
features = df[["Quantity", "Price"]].apply(
    pd.to_numeric, errors="coerce"
)

# Calculate the correlation matrix
correlation_matrix = features.corr(method="pearson")

print("Correlation Matrix:")
display(correlation_matrix)

# Display the correlation value
correlation_value = correlation_matrix.loc["Quantity", "Price"]

print("\nCorrelation between Quantity and Price:",
      correlation_value)

# Interpret the direction of correlation
if pd.isna(correlation_value):
    print("Correlation cannot be calculated.")
elif correlation_value > 0:
    print("The correlation is positive.")
elif correlation_value < 0:
    print("The correlation is negative.")
else:
    print("The correlation is zero.")

# Create the correlation heatmap
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```


## Question 10 – What Is Multicollinearity?

### Coding Exercise:

Using the `online_retail_II.xlsx` dataset:

1. Select the numerical features `Quantity` and `Price`.
2. Calculate the correlation matrix.
3. Identify feature pairs with an absolute correlation greater than 0.8.
4. Display the highly correlated feature pairs, if any.
5. Print a note about checking multicollinearity before building a Machine Learning model.

Do not assume that correlation alone proves multicollinearity or causation.


```python

import pandas as pd

# Load the dataset
df = pd.read_excel(
    "online_retail_II.xlsx",
    sheet_name="Year 2009-2010"
)

# Select numerical features
features = df[["Quantity", "Price"]].apply(
    pd.to_numeric, errors="coerce"
)

# Calculate the correlation matrix
correlation_matrix = features.corr()

print("Correlation Matrix:")
display(correlation_matrix)

# Identify highly correlated feature pairs
threshold = 0.8
high_correlation_pairs = []

columns = correlation_matrix.columns

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        correlation_value = correlation_matrix.iloc[i, j]

        if (
            pd.notna(correlation_value)
            and abs(correlation_value) > threshold
        ):
            high_correlation_pairs.append(
                (
                    columns[i],
                    columns[j],
                    correlation_value
                )
            )

# Display the results
print("\nHighly Correlated Feature Pairs:")

if high_correlation_pairs:
    for feature1, feature2, value in high_correlation_pairs:
        print(
            f"{feature1} and {feature2}: "
            f"Correlation = {value:.3f}"
        )
else:
    print("No feature pairs exceeded the threshold of 0.8.")

print(
    "\nNote: High correlation can indicate potential "
    "multicollinearity. For a more complete assessment, "
    "use methods such as Variance Inflation Factor (VIF) "
    "with an appropriate set of predictor variables."
)
```


## Question 12 – Why Should Outliers Not Always Be Removed?

### Coding Exercise:

Using the `online_retail_II.xlsx` dataset:

1. Select the `Quantity` column.
2. Detect outliers using the IQR method.
3. Display the total number of records and outlier records.
4. Display a sample of the detected outliers.
5. Create a box plot to visualize the outliers.
6. Compare the original dataset shape with the dataset shape after analysis.

Do not remove or modify any outliers. Treat extreme values as records that require investigation.


```python

import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_excel(
    "online_retail_II.xlsx",
    sheet_name="Year 2009-2010"
)

original_shape = df.shape

# Select the Quantity column
quantity = pd.to_numeric(df["Quantity"], errors="coerce")

# Calculate IQR bounds
q1 = quantity.quantile(0.25)
q3 = quantity.quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# Identify outliers
outlier_mask = (
    (quantity < lower_bound) |
    (quantity > upper_bound)
)

outliers = df[outlier_mask]

# Display results
print("Total Records:", len(df))
print("Number of Outliers:", len(outliers))
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nSample Outlier Records:")
display(outliers.head(10))

# Create a box plot
plt.figure(figsize=(8, 4))
plt.boxplot(quantity.dropna(), vert=False)
plt.title("Quantity Outlier Analysis")
plt.xlabel("Quantity")
plt.tight_layout()
plt.show()

# Confirm that the dataset has not been modified
print("\nOriginal Dataset Shape:", original_shape)
print("Dataset Shape After Analysis:", df.shape)
print("Dataset Unchanged:", original_shape == df.shape)

print(
    "\nConclusion: Outliers may represent valid transactions, "
    "bulk purchases, or product returns. Investigate them "
    "before deciding whether any action is necessary."
)
```


## Question 14 – Generate a Complete EDA Summary

### Coding Exercise:

Using the `online_retail_II.xlsx` dataset:

1. Display the dataset dimensions and column names.
2. Calculate missing values and duplicate records.
3. Display descriptive statistics for numerical features.
4. Count unique values in categorical columns.
5. Detect outliers in `Quantity` and `Price` using the IQR method.
6. Calculate the correlation between `Quantity` and `Price`.
7. Create a summary report containing the main data quality and numerical analysis results.

Do not delete records, fill missing values, or remove outliers.


```python

import pandas as pd

# Load the dataset
df = pd.read_excel(
    "online_retail_II.xlsx",
    sheet_name="Year 2009-2010"
)

# 1. Dataset overview
print("DATASET OVERVIEW")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn Names:")
print(df.columns.tolist())

# 2. Data quality checks
missing_count = df.isnull().sum().sum()
duplicate_count = df.duplicated().sum()

print("\nDATA QUALITY")
print("Total Missing Values:", missing_count)
print("Total Duplicate Records:", duplicate_count)

# 3. Numerical descriptive statistics
numeric_df = df.select_dtypes(include="number")

print("\nDESCRIPTIVE STATISTICS")
display(numeric_df.describe())

# 4. Categorical unique values
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

print("\nUNIQUE CATEGORICAL VALUES")

for column in categorical_columns:
    print(f"{column}: {df[column].nunique()}")

# 5. IQR outlier detection
outlier_counts = {}

for column in ["Quantity", "Price"]:
    values = pd.to_numeric(
        df[column], errors="coerce"
    ).dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = (
        (values < lower_bound) |
        (values > upper_bound)
    ).sum()

    outlier_counts[column] = outlier_count

print("\nIQR OUTLIER COUNTS")
for column, count in outlier_counts.items():
    print(f"{column}: {count}")

# 6. Correlation analysis
features = df[["Quantity", "Price"]].apply(
    pd.to_numeric, errors="coerce"
)

correlation = features.corr().loc["Quantity", "Price"]

print("\nCORRELATION")
print("Quantity vs Price:", correlation)

# 7. Generate a summary report
summary_report = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Total Missing Values",
        "Total Duplicate Records",
        "Quantity Outliers (IQR)",
        "Price Outliers (IQR)",
        "Quantity-Price Correlation"
    ],
    "Result": [
        df.shape[0],
        df.shape[1],
        missing_count,
        duplicate_count,
        outlier_counts["Quantity"],
        outlier_counts["Price"],
        correlation
    ]
})

print("\nCOMPLETE EDA SUMMARY")
display(summary_report)
```
